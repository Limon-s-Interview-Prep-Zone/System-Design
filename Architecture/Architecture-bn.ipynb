# সফটওয়্যার আর্কিটেকচার গাইড: সিস্টেম আর্কিটেকচার, অ্যাপ্লিকেশন আর্কিটেকচার এবং আর্কিটেকচারাল প্যাটার্নস

সফটওয়্যার আর্কিটেকচার বর্ণনা করে কীভাবে একটি সফটওয়্যার সিস্টেম গঠিত হয়, কীভাবে এর উপাদানগুলো একে অপরের সাথে ক্রিয়া-প্রতিক্রিয়া করে, এবং কীভাবে নকশা সংক্রান্ত সিদ্ধান্তগুলো স্কেলেবিলিটি (প্রসারণযোগ্যতা), রক্ষণাবেক্ষণযোগ্যতা, নিরাপত্তা, কর্মক্ষমতা ও নির্ভরযোগ্যতার মতো প্রয়োজনীয় বিষয়গুলোকে সমর্থন করে।

1. **সিস্টেম আর্কিটেকচার (Macro-Level)**: সম্পূর্ণ সিস্টেমটি কীভাবে গঠিত ও সংযুক্ত তা বর্ণনা করে—সার্ভার, নেটওয়ার্ক, ক্লাউড ডিপ্লয়মেন্ট এবং অ্যাপ্লিকেশনগুলোর মধ্যকার যোগাযোগ। উদাহরণ: Monolith, Microservices, Distributed Systems।
2. **অ্যাপ্লিকেশন আর্কিটেকচার (Meso-Level)**: একটি নির্দিষ্ট অ্যাপ্লিকেশন বা সার্ভিসের ভেতরে সোর্স কোড কীভাবে সাজানো হবে তা নির্ধারণ করে—লেয়ার, মডিউল ও ডিপেন্ডেন্সির প্রবাহ। উদাহরণ: Layered Architecture, Clean Architecture, Modular Monolith, Microkernel।
3. **আর্কিটেকচারাল প্যাটার্নস ও মেকানিজম (Micro/Tactical-Level)**: সফটওয়্যার তৈরির সাধারণ সমস্যাগুলো সমাধানের নির্দিষ্ট কৌশল—ডেটা কনসিস্টেন্সি, যোগাযোগ, পারফরম্যান্স এবং ব্যর্থতা নিয়ন্ত্রণ। উদাহরণ: CQRS, Event Sourcing, Outbox Pattern, Circuit Breaker, Mediator।

## পর্ব ১: সিস্টেম আর্কিটেকচার (System Architecture - Macro)

সিস্টেম আর্কিটেকচার সম্পূর্ণ এন্টারপ্রাইজ জুড়ে মূল ডিপ্লয়মেন্ট টপোলজি, ফিজিক্যাল সীমানা এবং সার্ভিসের মধ্যকার যোগাযোগ পরিচালনা করে।

### ১.১ মনোলিথিক বনাম ডিস্ট্রিবিউটেড স্পেকট্রাম (The Monolithic vs. Distributed Spectrum)
সিস্টেমের সমগ্র অ্যাপ্লিকেশন কার্যকারিতা একটিমাত্র অ্যাপ্লিকেশন ইউনিট হিসেবে প্যাকেজ ও ডিপ্লয় করা হয়।
সুবিধাসমূহ (Advantages)
- সহজ ডিপ্লয়মেন্ট এবং ডিবাগিং।
- ইন-প্রসেস মেথড কল অবিশ্বাস্য দ্রুতগতির।
- ডেটাবেস ট্রানজ্যাকশন ও ACID কনসিস্টেন্সি রক্ষা করা সহজ।
- অবকাঠামো ও অপারেশনাল জটিলতা ন্যূনতম।
  
ট্রেড-অফ ও সীমাবদ্ধতা (Trade-offs)
- একটি নির্দিষ্ট ফিচারকে স্বাধীনভাবে স্কেল করা কঠিন।
- একটি একক শেয়ার্ড রিলিজ পুরো অ্যাপ্লিকেশনের ওপর প্রভাব ফেলে।
- বাউন্ডারি সতর্কতার সাথে ডিজাইন না করলে মডিউলগুলোর মধ্যে টাইট কাপলিং তৈরি হয়।
- একাধিক বৃহৎ টিম একই কোডবেসে কাজ করলে মার্জ কনফ্লিক্ট বাড়ে।

সিস্টেম আর্কিটেকচারের সবচেয়ে প্রাথমিক সিদ্ধান্ত হলো **সিঙ্গেল-প্রসেস মনোলিথ** এবং **ডিস্ট্রিবিউটেড নেটওয়ার্ক আর্কিটেকচার**-এর মধ্যে সঠিকটি বেছে নেওয়া:
- **মনোলিথিক সিস্টেম**: একটিমাত্র ওএস প্রসেস মেমোরি স্পেসে চলে (বা লোড ব্যালেন্সারের পেছনে হুবহু রেপ্লিকেটেড ক্লাস্টার)। ইন-প্রসেস মেথড কলগুলো ন্যানোসেকেন্ড/মাইক্রোসেকেন্ডে কোনো নেটওয়ার্ক ওভারহেড ছাড়াই সম্পন্ন হয় এবং একক রিলেশনাল ডেটাবেসে নেটিভ ACID ট্রানজ্যাকশন সমর্থন করে।
- **ডিস্ট্রিবিউটেড সিস্টেম**: বিভিন্ন ফাংশনালিটি একাধিক স্বাধীন নেটওয়ার্ক নোডে বিভক্ত থাকে। এটি স্বাধীন স্কেলিং ও ফল্ট আইসোলেশন দেয়, কিন্তু নেটওয়ার্ক লেটেন্সি, আংশিক ব্যর্থতা (Partial Failures) এবং ইভেনচুয়াল কনসিস্টেন্সির ঝুঁকি তৈরি করে (*Fallacies of Distributed Computing*)।

### ১.২ মাইক্রোসার্ভিসেস আর্কিটেকচার (Microservices Architecture)

মাইক্রোসার্ভিসেস আর্কিটেকচার একটি সিস্টেমকে কতগুলো ক্ষুদ্র, স্বায়ত্তশাসিত এবং শিথিলভাবে সংযুক্ত সার্ভিসে বিভক্ত করে। প্রতিটি সার্ভিস একটি নির্দিষ্ট ব্যবসায়িক সক্ষমতাকে কেন্দ্র করে গঠিত হয় এবং নিজস্ব স্টোরেজ কঠোরভাবে পরিচালনা করে (**Database-per-Service**)।

```
                      ┌─────────────────────────┐
                      │   API Gateway (YARP)    │
                      └────────────┬────────────┘
                                   │
         ┌─────────────────────────┼─────────────────────────┐
         ▼                         ▼                         ▼
┌──────────────────┐      ┌──────────────────┐      ┌──────────────────┐
│  Orders Service  │      │ Payments Service │      │ Shipping Service │
│                  │      │                  │      │                  │
│ ┌──────────────┐ │      │ ┌──────────────┐ │      │ ┌──────────────┐ │
│ │ Orders DB    │ │      │ │ Payments DB  │ │      │ │ Shipping DB  │ │
│ └──────────────┘ │      │ └──────────────┘ │      │ └──────────────┘ │
└──────────────────┘      └──────────────────┘      └──────────────────┘
```

#### সুবিধাসমূহ (Benefits)
- **সাংগঠনিক স্বাধীনতা (Conway's Law Alignment)**: ছোট ক্রস-ফাংশনাল টিম অন্য টিমকে ব্লক না করে স্বাধীনভাবে সার্ভিস ডেভেলপ, টেস্ট ও রিলিজ করতে পারে।
- **টার্গেটেড ইলাস্টিক স্কেলিং**: সম্পূর্ণ অ্যাপ্লিকেশন স্কেল না করে শুধু উচ্চ চাপযুক্ত সার্ভিস (যেমন Search বা Checkout) স্কেল করা যায়।
- **ফল্ট আইসোলেশন (Fault Isolation)**: রেকমেন্ডেশন সার্ভিসে মেমোরি লিক বা ক্র্যাশ ঘটলেও মূল পেমেন্ট বা অর্ডার সার্ভিস সচল থাকে।

#### অসুবিধা ও ট্রেড-অফ (Drawbacks & Trade-offs)
- **নেটওয়ার্ক লেটেন্সি ও নেটওয়ার্কের মিথ্যা ধারণা**: ইন-মেমোরি মেথড কল ১০-৫০ মিলিসেকেন্ডের নেটওয়ার্ক রাউন্ড-ট্রিপে পরিণত হয়।
- **ডিস্ট্রিবিউটেড ডেটা কনসিস্টেন্সি**: মাল্টি-ডেটাবেস ACID ট্রানজ্যাকশন অসম্ভব; সাগা প্যাটার্ন এবং ইভেনচুয়াল কনসিস্টেন্সি বাধ্যতামূলক।
- **বিশাল অপারেশনাল খরচ**: কুবারনেটিস ক্লাস্টার, এপিআই গেটওয়ে, ওপেনটেলিমেট্রি ট্রেসিং এবং সেন্ট্রালাইজড লগিং পরিচালনার বিশাল ওভারহেড তৈরি হয়।

#### টার্গেট টেকনোলজি ও .NET / C# NuGet প্যাকেজ
- `Yarp.ReverseProxy`: হাই-পারফরম্যান্স API Gateway তৈরির জন্য মাইক্রোসফটের রিভার্স প্রক্সি।
- `Microsoft.Extensions.Http.Resilience` / `Polly`: রেজিলিয়েন্ট HTTP পাইপলাইন (Retries, Circuit Breaker, Rate Limiting)।
- `Aspire.Hosting`: ডিস্ট্রিবিউটেড অ্যাপ্লিকেশন অর্কেস্ট্রেশন ও টেলিমেট্রির জন্য .NET Aspire।

#### মিনিমাল প্রোডাকশন C# কোড উদাহরণ

````python
// Production .NET 9 API Gateway configuration using YARP & Resilient HTTP Client
using Microsoft.AspNetCore.Builder;
using Microsoft.Extensions.DependencyInjection;
using Polly;
using Polly.Retry;

var builder = WebApplication.CreateBuilder(args);

// 1. Configure YARP Reverse Proxy
builder.Services.AddReverseProxy()
    .LoadFromConfig(builder.Configuration.GetSection("ReverseProxy"));

// 2. Configure Resilient Outbound HttpClient with Polly Circuit Breaker & Jitter Retry
builder.Services.AddHttpClient("PaymentService", client =>
{
    client.BaseAddress = new Uri("https://payment-service.internal/");
    client.Timeout = TimeSpan.FromSeconds(5);
})
.AddStandardResilienceHandler(options =>
{
    options.Retry.MaxRetryAttempts = 3;
    options.Retry.BackoffType = DelayBackoffType.Exponential;
    options.Retry.UseJitter = true;

    options.CircuitBreaker.SamplingDuration = TimeSpan.FromSeconds(10);
    options.CircuitBreaker.FailureRatio = 0.5; // Trip if 50% requests fail
    options.CircuitBreaker.BreakDuration = TimeSpan.FromSeconds(30);
});

var app = builder.Build();
app.MapReverseProxy();
app.Run();
````

### ১.৩ স্পেস-বেসড আর্কিটেকচার (Space-Based Architecture)

স্পেস-বেসড আর্কিটেকচার মূলত ডেভিড গেলার্নটারের টাপল স্পেস (Tuple Space) ধারণার উপর প্রতিষ্ঠিত। এর মূল লক্ষ্য হলো **সেন্ট্রাল ডেটাবেসের সিঙ্ক্রোনাস বটলনেক সম্পূর্ণ দূর করে** চরম স্কেলেবিলিটি, অতি-উচ্চ কনকারেন্সি এবং সাব-মিলিসেকেন্ড রেসপন্স টাইম নিশ্চিত করা।

```
                         ┌───────────────────────┐
                         │    Virtual Request    │
                         │    Director (Load)    │
                         └───────────┬───────────┘
                                     │
           ┌─────────────────────────┴─────────────────────────┐
           ▼                                                   ▼
┌─────────────────────────┐                         ┌─────────────────────────┐
│ Processing Unit 1 (RAM) │                         │ Processing Unit 2 (RAM) │
│                         │   Replication Engine    │                         │
│ • Business Logic        │◄───────────────────────►│ • Business Logic        │
│ • In-Memory Data Grid   │       (Sync/Async)      │ • In-Memory Data Grid   │
└──────────┬──────────────┘                         └───────────┬─────────────┘
           │                                                    │
           └──────────────────────────┬─────────────────────────┘
                                      ▼
                        ┌───────────────────────────┐
                        │ Async Data Writer (Batch) │
                        └─────────────┬─────────────┘
                                      ▼
                        ┌───────────────────────────┐
                        │   Persistent Database     │
                        └───────────────────────────┘
```

#### সুবিধাসমূহ (Benefits)
- **সাব-মিলিসেকেন্ড আল্ট্রা-লো লেটেন্সি**: সমস্ত রিড ও রাইট র‍্যামের (RAM) মধ্যে সম্পন্ন হওয়ায় ট্রানজ্যাকশন অবিশ্বাস্য গতিতে সম্পন্ন হয়।
- **ইলাস্টিক লোড শোষণ**: ডেটাবেস রো-লকিং বা ডেডলক ছাড়াই হঠাৎ আসা ১০০ গুণ ট্রাফিক স্পাইক ইন-মেমোরি প্রসেসিং ইউনিট স্কেল করে সামাল দেওয়া যায়।

#### অসুবিধা ও ট্রেড-অফ (Drawbacks & Trade-offs)
- **উচ্চ মেমোরি অবকাঠামো ব্যয়**: প্রচুর ডেটা দ্রুতগতির র‍্যামে ধরে রাখা এসএসডি স্টোরেজের চেয়ে অনেক বেশি ব্যয়বহুল।
- **ক্লাস্টার বিপর্যয়ে ডেটা লসের ঝুঁকি**: মেমোরির ডেটা ডিস্কে সিঙ্ক হওয়ার আগেই অপ্রত্যাশিত পাওয়ার ফেইলিউর হলে ডেটা হারানোর ঝুঁকি থাকে।

#### টার্গেট টেকনোলজি ও .NET / C# NuGet প্যাকেজ
- `Microsoft.Orleans`: ভার্চুয়াল অ্যাক্টর ফ্রেমওয়ার্ক (ডিস্ট্রিবিউটেড ইন-মেমোরি স্টেটফুল গ্রেইন ও অ্যাসিঙ্ক পারসিস্টেন্স)।
- `StackExchange.Redis`: ইন-মেমোরি ক্যাশিং এবং অ্যাটমিক অপারেশন।

#### মিনিমাল প্রোডাকশন C# কোড উদাহরণ

````python
// Space-Based Virtual Actor Pattern using Microsoft Orleans in .NET 9
using Orleans;

public interface ITradingAccountGrain : IGrainWithGuidKey
{
    Task<decimal> GetBalanceAsync();
    Task<bool> ExecuteOrderAsync(decimal orderAmount);
}

public class TradingAccountGrain : Grain, ITradingAccountGrain
{
    private readonly IPersistentState<AccountState> _state;

    public TradingAccountGrain(
        [PersistentState("tradingAccount", "accountStorage")] IPersistentState<AccountState> state)
    {
        _state = state;
    }

    public Task<decimal> GetBalanceAsync() => Task.FromResult(_state.State.Balance);

    public async Task<bool> ExecuteOrderAsync(decimal orderAmount)
    {
        // Executes in sub-millisecond RAM space with single-threaded actor guarantee
        if (_state.State.Balance < orderAmount)
            return false;

        _state.State.Balance -= orderAmount;
        _state.State.ExecutionCount++;

        // Asynchronously persists state to backing store without blocking caller
        await _state.WriteStateAsync();
        return true;
    }
}

public class AccountState
{
    public decimal Balance { get; set; } = 10_000m;
    public int ExecutionCount { get; set; }
}
````

### ১.৪ এন-টিয়ার আর্কিটেকচার (N-Tier Architecture - ফিজিক্যাল ডিপ্লয়মেন্ট টপোলজি)

**এন-টিয়ার আর্কিটেকচার (N-Tier Architecture)** হলো একটি **সিস্টেম আর্কিটেকচার (Macro-Level)** প্যাটার্ন যা একটি অ্যাপ্লিকেশনের এক্সিকিউশনকে **$N$ সংখ্যক পৃথক হার্ডওয়্যার নোড, সার্ভার বা নেটওয়ার্ক বাউন্ডারিতে** শারীরিকভাবে (physically) বিভক্ত করে।

```
┌─────────────────────────────────┐
│ Tier 1: Client Tier (Browser)   │
└────────────────┬────────────────┘
                 │ (Public WAN / HTTPS)
                 ▼
┌─────────────────────────────────┐
│ Tier 2: Web Server / DMZ        │  (Reverse Proxy / Edge TLS)
└────────────────┬────────────────┘
                 │ (Internal LAN / gRPC)
                 ▼
┌─────────────────────────────────┐
│ Tier 3: App Server Tier (Nodes) │  (Business Compute & Services)
└────────────────┬────────────────┘
                 │ (Private VLAN / TCP)
                 ▼
┌─────────────────────────────────┐
│ Tier 4: Database Tier (Cluster) │  (SQL / NoSQL Storage Cluster)
└─────────────────────────────────┘
```

#### সুবিধাসমূহ (Benefits)
- **নেটওয়ার্ক সিকিউরিটি ও DMZ আইসোলেশন**: কেবলমাত্র Tier 2 (Web/Reverse Proxy) পাবলিক ইন্টারনেটের সাথে যুক্ত থাকে। ডেটাবেস (Tier 4) এবং কোর অ্যাপ্লিকেশন সার্ভার (Tier 3) প্রাইভেট, বিচ্ছিন্ন VLAN-এ সুরক্ষিত থাকে যার কোনো পাবলিক আইপি থাকে না।
- **স্বাধীন হার্ডওয়্যার স্কেলিং**: কম্পিউট-হেভি অ্যাপ্লিকেশন স্তর CPU-অপ্টিমাইজড ক্লাউড নোডে স্কেল করা যায়, অন্যদিকে ডেটাবেস ক্লাস্টার হাই-IOPS স্টোরেজ মেশিনে স্কেল করা যায়।
- **ডিফেন্স ইন ডেপথ (Defense in Depth)**: পাবলিক ওয়েব সার্ভার হ্যাক হলেও আক্রমণকারী সরাসরি ডেটাবেসে প্রবেশ করতে পারে না।

#### অসুবিধা ও ট্রেড-অফ (Drawbacks & Trade-offs)
- **নেটওয়ার্ক লেটেন্সি ট্যাক্স**: প্রতিবার টিয়ার পার হতে নেটওয়ার্ক সকেট সিরিয়ালাইজেশন, ডিসিরিয়ালাইজেশন এবং ১০-৫০ মিলিসেকেন্ড লেটেন্সি যুক্ত হয়।
- **অপারেশনাল জটিলতা**: লোড ব্যালেন্সার, ফায়ারওয়াল রাউটিং টেবিল এবং টিয়ারগুলোর মধ্যে SSL/mTLS সার্টিফিকেট কনফিগার ও মনিটর করার বিশাল ওভারহেড তৈরি হয়।

#### টার্গেট টেকনোলজি ও .NET / C# NuGet প্যাকেজ
- `Yarp.ReverseProxy`: Tier 2 $\rightarrow$ Tier 3 এজ রাউটিং, টিএলএস টার্মিনেশন এবং লোড ব্যালেন্সিংয়ের জন্য।
- `Microsoft.AspNetCore.Authentication.JwtBearer`: ডিস্ট্রিবিউটেড JWT বা mTLS সার্টিফিকেটের মাধ্যমে টিয়ারগুলোর মধ্যকার যোগাযোগ সুরক্ষিত করার জন্য।

### ১.৫ সার্ভিস-ওরিয়েন্টেড আর্কিটেকচার (SOA - Service-Oriented Architecture)

**সার্ভিস-ওরিয়েন্টেড আর্কিটেকচার (SOA)** হলো ২০০০-এর দশকের একটি অত্যন্ত জনপ্রিয় এন্টারপ্রাইজ সিস্টেম টপোলজি, যা একটি **এন্টারপ্রাইজ সার্ভিস বাস (Enterprise Service Bus - ESB)** ব্যবহার করে প্রতিষ্ঠানের বিভিন্ন বৃহৎ ও স্বয়ংসম্পূর্ণ বিজনেস অ্যাপ্লিকেশনগুলোকে সমন্বিত করে।

```
┌─────────────────┐       ┌─────────────────┐       ┌─────────────────┐
│ Billing Service │       │   CRM Service   │       │   ERP Service   │
└────────┬────────┘       └────────┬────────┘       └────────┬────────┘
         │ (SOAP/XML)              │ (WSDL)                  │ (Binary)
         ▼                         ▼                         ▼
┌─────────────────────────────────────────────────────────────────────┐
│                   Enterprise Service Bus (ESB)                      │
│   • Message Transformation  • Protocol Mediation  • Routing Logic   │
└──────────────────────────────────┬──────────────────────────────────┘
                                   ▼
                       Shared Enterprise Database
```

#### সবচেয়ে মৌলিক তুলনা: SOA বনাম মাইক্রোসার্ভিসেস (SOA vs. Microservices)
- **SOA দর্শন**: *"স্মার্ট পাইপস, ডাম্ব এন্ডপয়েন্টস" (Smart pipes, dumb endpoints)*। কেন্দ্রীয় ESB-এর ভেতরে জটিল বিজনেস রাউটিং, প্রোটোকল রূপান্তর (SOAP থেকে REST, XML থেকে JSON) এবং মেসেজ মিডিয়েশন সম্পন্ন হয়। সার্ভিসগুলো প্রায়শই একটি বিশাল শেয়ার্ড এন্টারপ্রাইজ ডেটাবেস ব্যবহার করে।
- **মাইক্রোসার্ভিসেস দর্শন**: *"ডাম্ব পাইপস, স্মার্ট এন্ডপয়েন্টস" (Dumb pipes, smart endpoints)*। যোগাযোগের মাধ্যম (যেমন: RabbitMQ, Kafka, HTTP) কোনো বিজনেস লজিক ধারণ করে না। সার্ভিসগুলো অত্যন্ত সূক্ষ্ম ও স্বায়ত্তশাসিত এবং নিজস্ব প্রাইভেট ডেটাবেসের কঠোর মালিকানা বজায় রাখে।

#### সুবিধাসমূহ (Benefits)
- **লিগ্যাসি এন্টারপ্রাইজ সিস্টেম ইন্টিগ্রেশন**: পুরনো মেইনফ্রেম, SAP/ERP সিস্টেম এবং আধুনিক ওয়েব সার্ভিসগুলোকে কোনো কোড পরিবর্তন ছাড়াই অনায়াসে যুক্ত করতে সক্ষম।
- **প্রোটোকল ও ডেটা ফরম্যাট রূপান্তর**: বিভিন্ন সিস্টেমের প্রোটোকল অমিল (যেমন: লিগ্যাসি XML এবং আধুনিক প্রোটোকল) ESB নিজে রূপান্তর করে সমাধান করে।

#### অসুবিধা ও ট্রেড-অফ (Drawbacks & Trade-offs)
- **সিঙ্গেল পয়েন্ট অব ফেইলিউর ও বটলনেক**: ESB নিজেই একটি দানবীয়, জটিল মনোলিথে পরিণত হয়; ESB ক্র্যাশ করলে সম্পূর্ণ এন্টারপ্রাইজ অচল হয়ে পড়ে।
- **জটিল ভেন্ডর নির্ভরতা**: অত্যন্ত ব্যয়বহুল প্রোপ্রাইটরি ESB সফটওয়্যার (BizTalk, IBM WebSphere, TIBCO) এবং লাইসেন্সিং খরচের উপর অতি-নির্ভরশীলতা।

#### টার্গেট টেকনোলজি ও .NET / C# NuGet প্যাকেজ
- `CoreWCF`: এন্টারপ্রাইজ SOAP/XML ইন্টিগ্রেশনের জন্য আধুনিক .NET 8/9 সমর্থিত Windows Communication Foundation (WCF) পোর্ট।
- `MassTransit`: মেসেজ-ভিত্তিক এন্টারপ্রাইজ ইন্টিগ্রেশনের জন্য ভারী ESB-র আধুনিক ও হালকা বিকল্প।

### ১.৬ সার্ভারলেস আর্কিটেকচার (Serverless - FaaS / BaaS Architecture)

**সার্ভারলেস আর্কিটেকচার (Serverless Architecture)** হলো একটি ক্লাউড-নেটিভ সিস্টেম টপোলজি যেখানে ক্লাউড প্রোভাইডার সম্পূর্ণভাবে ডায়নামিক কন্টেইনার বরাদ্দ, অটো-স্কেলিং এবং হার্ডওয়্যার প্রভিশনিং পরিচালনা করে। কম্পিউট লজিক স্বল্পস্থায়ী ও ইভেন্ট-ট্রিগার ফাংশন হিসেবে কার্যকর হয় (**Function-as-a-Service - FaaS**), যা ক্লাউডের ম্যানেজড স্টোরেজ ও ডেটাবেস দ্বারা সমর্থিত (**Backend-as-a-Service - BaaS**)।

```
┌─────────────────────────────────┐
│ Event Trigger (HTTP, Queue, S3) │
└────────────────┬────────────────┘
                 │ (Event Invocations)
                 ▼
┌────────────────────────────────────────────────────────┐
│             Ephemeral Compute Cluster (FaaS)           │
│                                                        │
│  ┌──────────────────┐ ┌──────────────────┐ ┌─────────┐ │
│  │ Function Instance│ │ Function Instance│ │ ...     │ │
│  │ (Spins up on-load) │ (Spins up on-load) │         │ │
│  └────────┬─────────┘ └────────┬─────────┘ └────┬────┘ │
└───────────┼────────────────────┼────────────────┼──────┘
            │                    │                │
            └────────────────────┼────────────────┘
                                 ▼
┌────────────────────────────────────────────────────────┐
│             Managed Cloud Backends (BaaS)              │
│  • DynamoDB / Cosmos DB  • Blob Storage  • Event Grid  │
└────────────────────────────────────────────────────────┘
```

#### সুবিধাসমূহ (Benefits)
- **শূন্য নিষ্ক্রিয় খরচ (Zero Idle Cost)**: অ্যাপ্লিকেশন যখন কোনো রিকোয়েস্ট প্রসেস করে না তখন অবকাঠামো খরচ একদম শূন্য (\$0); শুধুমাত্র এক্সিকিউশন সময়ের মিলিসেকেন্ডের জন্য অর্থ প্রদান করতে হয়।
- **তাত্ক্ষণিক ইলাস্টিক অটো-স্কেলিং**: ট্রাফিকের হঠাৎ বন্যায় কয়েক সেকেন্ডের মধ্যে শূন্য থেকে ১০,০০০ সমান্তরাল ফাংশন ইন্সটেন্সে স্কেল হতে পারে।
- **সার্ভার ও ওএস ব্যবস্থাপনামুক্ত**: ওএস প্যাচিং, সিকিউরিটি আপডেট বা নোড ক্লাস্টার পরিচালনার কোনো ঝামেলা থাকে না।

#### অসুবিধা ও ট্রেড-অফ (Drawbacks & Trade-offs)
- **কোল্ড স্টার্ট লেটেন্সি (Cold Start Latency)**: নিষ্ক্রিয় ফাংশন প্রথমবার রিকোয়েস্ট পেলে নতুন মাইক্রোভিম (microVM) বা কন্টেইনার চালু করতে ১০০ মিলিসেকেন্ড থেকে ২ সেকেন্ড পর্যন্ত লেটেন্সি হতে পারে।
- **স্টেটলেস সীমাবদ্ধতা**: মেমোরির ভেতরে রিকোয়েস্টের মধ্যবর্তী কোনো স্টেট ধরে রাখা যায় না; প্রতিটি স্টেট এক্সটার্নাল স্টোরেজে (Redis, Cosmos DB) রাখতে ও রিড করতে হয়।
- **ভেন্ডর লক-ইন (Vendor Lock-In)**: ক্লাউড-নির্দিষ্ট ট্রিগারের সাথে গভীর সংযোগ (AWS Lambda + SQS বনাম Azure Functions + Service Bus)।

#### টার্গেট টেকনোলজি ও .NET / C# NuGet প্যাকেজ
- `Microsoft.Azure.Functions.Worker`: Azure Functions-এর জন্য .NET Isolated Worker প্রসেস মডেল।
- `Amazon.Lambda.AspNetCoreServer`: AWS Lambda-তে সরাসরি ASP.NET Core Minimal APIs চালানোর জন্য।

#### মিনিমাল প্রোডাকশন C# কোড উদাহরণ

### ১.৭ সেল-বেসড আর্কিটেকচার (Cell-Based / Cellular Architecture)

**সেল-বেসড আর্কিটেকচার (Cell-Based Architecture)** হলো একটি আধুনিক ও উচ্চমানের **সিস্টেম আর্কিটেকচার** টপোলজি যা বিশ্বের শীর্ষস্থানীয় হাইপারস্কেল ক্লাউড প্ল্যাটফর্মগুলো (যেমন: **AWS Route 53**, **Slack Workspaces**, **Netflix**, এবং **Stripe**) তাদের সিস্টেমে **ব্লাস্ট রেডিয়াস (Blast Radius) সম্পূর্ণ সীমাবদ্ধ করতে** এবং স্কেলিংয়ের চরম সীমাবদ্ধতা অতিক্রম করতে ব্যবহার করে।

```
                      ┌───────────────────────────────────────┐
                      │    Cell Router / Partition Gateway    │
                      └──────────────────┬────────────────────┘
                                         │
         ┌───────────────────────────────┼───────────────────────────────┐
         ▼                               ▼                               ▼
┌──────────────────┐            ┌──────────────────┐            ┌──────────────────┐
│  Cell 1 (100k)   │            │  Cell 2 (100k)   │            │  Cell 3 (100k)   │
│                  │            │                  │            │                  │
│ • App Instances  │            │ • App Instances  │            │ • App Instances  │
│ • Private DB     │            │ • Private DB     │            │ • Private DB     │
│ • Private Cache  │            │ • Private Cache  │            │ • Private Cache  │
│ (Users A - G)    │            │ (Users H - P)    │            │ (Users Q - Z)    │
└──────────────────┘            └──────────────────┘            └──────────────────┘
```

#### আর্কিটেকচার গঠন ও কর্মপদ্ধতি (Anatomy & Mechanics)
- **দ্য সেল (The Cell)**: সম্পূর্ণ সিস্টেমের একটি স্বয়ংসম্পূর্ণ ও স্বাধীন ডিপ্লয়মেন্ট (অ্যাপ্লিকেশন কম্পিউট, ডেটাবেস, ক্যাশ, মেসেজ কিউ)। প্রতিটি সেল একটি স্বতন্ত্র দ্বীপের মতো কাজ করে এবং **অন্য সেলের উপর এর কোনো নির্ভরতা থাকে না**।
- **দ্য সেল রাউটার (The Cell Router)**: অত্যন্ত নির্ভরযোগ্য একটি গেটওয়ে যা ইনকামিং রিকোয়েস্টের টেন্যান্ট বা ইউজার আইডি (যেমন: `TenantId`, `UserId`) পরীক্ষা করে সুনির্দিষ্ট সেলে ট্রাফিক পাঠিয়ে দেয়।
- **ব্লাস্ট রেডিয়াস সীমাবদ্ধকরণ (Blast-Radius Containment)**: কোনো মারাত্মক সফটওয়্যার বাগ বা ডেটাবেস ক্র্যাশে Cell 1 সম্পূর্ণ ধ্বংস হলেও, **Cell 2, Cell 3 এবং Cell 4 শতভাগ (100%) সচল থাকে**। সম্পূর্ণ ১০০% ইউজারের বদলে মাত্র ৫-১০% ইউজার প্রভাবিত হয়।

#### সুবিধাসমূহ (Benefits)
- **নিশ্চিত উচ্চ প্রাপ্যতা ও সীমিত ব্লাস্ট রেডিয়াস**: অপ্রত্যাশিত বিপর্যয় গাণিতিকভাবে নির্দিষ্ট ভগ্নাংশের গ্রাহকদের মধ্যেই সীমাবদ্ধ থাকে।
- **পূর্বাভাসযোগ্য পারফরম্যান্স (Predictable Performance)**: প্রতিটি সেলের সর্বোচ্চ আকার নির্দিষ্ট করে দেওয়ায় (যেমন: প্রতি সেলে সর্বোচ্চ ১ লাখ ইউজার) ডেটাবেস সাইজ বা কুয়েরি লেটেন্সি কখনো নিয়ন্ত্রণের বাইরে যায় না।
- **শূন্য-ঝুঁকির ক্যানারি রোলআউট**: নতুন সফটওয়্যার রিলিজ প্রথমে কেবল Cell 1-এ ডিপ্লয় করে নিরাপদে পরীক্ষা করা যায়।

#### অসুবিধা ও ট্রেড-অফ (Drawbacks & Trade-offs)
- **জটিল সেল রাউটার আর্কিটেকচার**: সেল রাউটারটিকে প্রায় অমর বা ধ্বংস-অপ্রতিরোধ্য হতে হয় (সহজ, সর্বোচ্চ প্রাপ্যতা সম্পন্ন এবং ন্যূনতম মুভিং পার্টস বিশিষ্ট)।
- **ক্রস-সেল অপারেশন ও ডেটা সিঙ্ক**: সেলগুলোর মধ্যকার যৌথ কুয়েরি, গ্লোবাল অ্যানালিটিক্স বা এক সেল থেকে অন্য সেলে গ্রাহক মাইগ্রেশনের জন্য বিশেষ ডেটা পাইপলাইনের প্রয়োজন হয়।

#### টার্গেট টেকনোলজি ও .NET / C# NuGet প্যাকেজ
- `Yarp.ReverseProxy`: উচ্চ-ক্ষমতাসম্পন্ন টেন্যান্ট-ভিত্তিক সেল রাউটিং রুলস বাস্তবায়নের জন্য।

### ২.১ লেয়ারড আর্কিটেকচার (Layered Architecture - লজিক্যাল বাউন্ডারি ও বিভাজন)

> [!IMPORTANT]
> **আর্কিটেকচারের সবচেয়ে গুরুত্বপূর্ণ পার্থক্য: লেয়ার বনাম টিয়ার (Layers vs. Tiers)**
> - **লেয়ার (Layers) = কোডের লজিক্যাল বিন্যাস**: প্রেজেন্টেশন, বিজনেস লজিক এবং ডেটা এক্সেসের মতো অংশগুলো **একটিমাত্র ওএস প্রসেস এবং মেমোরি স্পেসের ভেতরে** বিভক্ত থাকে। এদের যোগাযোগ ইন-মেমোরি মেথড কলের মাধ্যমে ন্যানোসেকেন্ডে ঘটে।
> - **টিয়ার (Tiers) = ফিজিক্যাল নেটওয়ার্ক ডিপ্লয়মেন্ট**: অ্যাপ্লিকেশনটি শারীরিকভাবে **পৃথক হার্ডওয়্যার মেশিন, ওএস প্রসেস এবং নেটওয়ার্ক সীমানায়** বিভক্ত থাকে যা সিরিয়ালাইজড নেটওয়ার্ক প্যাকেটের (HTTP, gRPC, TCP) মাধ্যমে যোগাযোগ করে।
> - *একটি ৩-লেয়ার (3-Layered) অ্যাপ্লিকেশন ১-টিয়ার (একক ডেস্কটপ/মনোলিথ) হিসেবেও ডিপ্লয় করা যায়, আবার ৩-টিয়ার (ব্রাউজার ক্লায়েন্ট $\rightarrow$ এপিআই সার্ভার $\rightarrow$ ডেটাবেস) হিসেবেও ডিস্ট্রিবিউট করা যায়।*

![Logical Layers vs Physical Tiers](images/layers-vs-tiers.svg)

লেয়ারড আর্কিটেকচার কোডবেসকে কতগুলো অনুভূমিক লজিক্যাল স্তরে ভাগ করে, যেখানে প্রতিটি স্তরের নির্দিষ্ট দায়িত্ব থাকে এবং প্রতিটি স্তর কেবল তার ঠিক নিচের স্তরের উপর নির্ভর করে (*Strict Layering*) অথবা নিচের যেকোনো স্তরের উপর নির্ভর করে (*Relaxed Layering*)।

```
┌─────────────────────────────────────────┐
│     Presentation Layer (Controllers)    │
└────────────────────┬────────────────────┘
                     ▼
┌─────────────────────────────────────────┐
│   Business Logic Layer (Services/BLL)   │
└────────────────────┬────────────────────┘
                     ▼
┌─────────────────────────────────────────┐
│    Data Access Layer (DAL/Repository)   │
└────────────────────┬────────────────────┘
                     ▼
┌─────────────────────────────────────────┐
│          Relational Database            │
└─────────────────────────────────────────┘
```

#### সুবিধাসমূহ (Benefits)
- **সহজবোধ্যতা ও সর্বজনীন পরিচিতি**: সফটওয়্যার ইঞ্জিনিয়ারিং টিমের কাছে সবচেয়ে পরিচিত ও সহজ প্যাটার্ন।
- **দায়িত্বের স্পষ্ট বিভাজন**: UI লজিক এবং ডেটাবেস এক্সেস কোডের মধ্যে সুস্পষ্ট সীমানা থাকে।
- **দ্রুত প্রাথমিক ডেভেলপমেন্ট**: সাধারণ CRUD এবং প্রশাসনিক অ্যাপ্লিকেশনের জন্য অত্যন্ত দ্রুত বাস্তবায়নযোগ্য।

#### অসুবিধা ও ট্রেড-অফ (Drawbacks & Trade-offs)
- **ডেটাবেসের সাথে শক্ত কাপলিং**: বিজনেস লেয়ারটি প্রায়শই পিওর ডোমেইন নিয়মের পরিবর্তে ডেটাবেস টেবিল ও সত্তার উপর নির্ভরশীল হয়ে পড়ে।
- **আর্কিটেকচার সিঙ্কহোল অ্যান্টি-প্যাটার্ন**: কোনো বিজনেস লজিক ছাড়াই রিকোয়েস্ট অপ্রয়োজনীয়ভাবে একাধিক স্তরের মধ্য দিয়ে পাস হয়, যা অপ্রয়োজনীয় বয়লারপ্লেট তৈরি করে।

#### টার্গেট টেকনোলজি ও .NET / C# NuGet প্যাকেজ
- `Microsoft.Extensions.DependencyInjection`: স্ট্যান্ডার্ড IoC কন্টেইনার।
- `Microsoft.EntityFrameworkCore`: ডেটা এক্সেসের জন্য স্ট্যান্ডার্ড ওআরএম।

#### মিনিমাল প্রোডাকশন C# কোড উদাহরণ
```csharp
// Layered Architecture: Presentation, BLL, এবং DAL-এর স্পষ্ট বিভাজন
using Microsoft.AspNetCore.Mvc;
using Microsoft.EntityFrameworkCore;

public record UserEntity(Guid Id, string Email, string FullName, bool IsActive);

public class AppDbContext(DbContextOptions<AppDbContext> options) : DbContext(options)
{
    public DbSet<UserEntity> Users => Set<UserEntity>();
}

public interface IUserRepository
{
    Task<UserEntity?> GetByIdAsync(Guid id, CancellationToken ct = default);
    Task AddAsync(UserEntity user, CancellationToken ct = default);
}

public class UserRepository(AppDbContext db) : IUserRepository
{
    public Task<UserEntity?> GetByIdAsync(Guid id, CancellationToken ct = default) =>
        db.Users.AsNoTracking().FirstOrDefaultAsync(u => u.Id == id, ct);

    public async Task AddAsync(UserEntity user, CancellationToken ct = default)
    {
        await db.Users.AddAsync(user, ct);
        await db.SaveChangesAsync(ct);
    }
}

public record CreateUserCommand(string Email, string FullName);
public record UserDto(Guid Id, string Email, string FullName);

public interface IUserService
{
    Task<UserDto> RegisterUserAsync(CreateUserCommand command, CancellationToken ct = default);
}

public class UserService(IUserRepository userRepository) : IUserService
{
    public async Task<UserDto> RegisterUserAsync(CreateUserCommand command, CancellationToken ct = default)
    {
        if (string.IsNullOrWhiteSpace(command.Email) || !command.Email.Contains('@'))
            throw new ArgumentException("Invalid email address format.");

        var entity = new UserEntity(Guid.NewGuid(), command.Email.Trim().ToLowerInvariant(), command.FullName, true);
        await userRepository.AddAsync(entity, ct);

        return new UserDto(entity.Id, entity.Email, entity.FullName);
    }
}
```

### ২.২ ক্লিন / হেক্সাগোনাল আর্কিটেকচার (Clean / Hexagonal Architecture)

ক্লিন আর্কিটেকচার **Dependency Inversion Principle (DIP)**-এর উপর ভিত্তি করে তৈরি: ব্যবসায়িক নিয়মাবলী থাকবে সিস্টেমের কেন্দ্রে এবং ডেটাবেস, ফ্রেমওয়ার্ক বা বাহ্যিক অবকাঠামোর উপর এর কোনো নির্ভরতা থাকবে না।

![Clean vs Layered Architecture](images/clean-vs-layered-architecture.svg)

#### সুবিধাসমূহ (Benefits)
- **প্রযুক্তি ও ডেটাবেস স্বাধীনতা**: ডেটাবেস বা ফ্রেমওয়ার্ক পরিবর্তন করলেও কোর ডোমেইন লজিকে কোনো হাত দিতে হয় না।
- **শতভাগ টেস্টেবিলিটি**: ডেটাবেস বা সার্ভার ছাড়াই ইন-মেমোরি মকের মাধ্যমে ইউজ কেস টেস্ট করা সম্ভব।
- **দীর্ঘমেয়াদী স্থায়িত্ব**: ব্যবসায়িক কোড থার্ড-পার্টি লাইব্রেরির পরিবর্তন থেকে নিরাপদ থাকে।

#### অসুবিধা ও ট্রেড-অফ (Drawbacks & Trade-offs)
- **উচ্চ ইনডাইরেকশন ও ফাইলের সংখ্যা বৃদ্ধি**: একটি সাধারণ ফিচারের জন্য একাধিক প্রজেক্টে বেশ কয়েকটি ফাইল তৈরি করতে হয়।
- **সহজ অ্যাপ্লিকেশনের জন্য অতিরিক্ত জটিলতা**: সাধারণ CRUD অ্যাপ্লিকেশনে এটি অপ্রয়োজনীয় স্থাপত্যিক জটিলতা আনে।

#### টার্গেট টেকনোলজি ও .NET / C# NuGet প্যাকেজ
- `MediatR`: ইন-প্রসেস ইউজ কেস ডিসপ্যাচ করার জন্য।
- `FluentValidation`: ইউজ কেস হ্যান্ডলারে পৌঁছানোর আগে ডিক্লারেটিভ ভ্যালিডেশন নিশ্চিত করার জন্য।

#### মিনিমাল প্রোডাকশন C# কোড উদাহরণ
```csharp
// Clean Architecture: Port-and-Adapter flow with MediatR & C# 12
using MediatR;

// --- Domain Core (কোনো ফ্রেমওয়ার্ক নির্ভরতা নেই) ---
public sealed class Order
{
    public Guid Id { get; private set; }
    public decimal TotalAmount { get; private set; }
    public bool IsConfirmed { get; private set; }

    private Order() { }

    public static Order Create(decimal amount)
    {
        if (amount <= 0) throw new ArgumentException("Order amount must be positive.");
        return new Order { Id = Guid.NewGuid(), TotalAmount = amount, IsConfirmed = false };
    }

    public void Confirm()
    {
        if (IsConfirmed) throw new InvalidOperationException("Order already confirmed.");
        IsConfirmed = true;
    }
}

// --- Application Layer (Use Cases & Ports) ---
public interface IOrderRepositoryPort
{
    Task<Order?> GetByIdAsync(Guid id, CancellationToken ct);
    Task SaveAsync(Order order, CancellationToken ct);
}

public record ConfirmOrderCommand(Guid OrderId) : IRequest<bool>;

public class ConfirmOrderCommandHandler(IOrderRepositoryPort orderRepo) 
    : IRequestHandler<ConfirmOrderCommand, bool>
{
    public async Task<bool> Handle(ConfirmOrderCommand request, CancellationToken ct)
    {
        var order = await orderRepo.GetByIdAsync(request.OrderId, ct)
            ?? throw new KeyNotFoundException($"Order {request.OrderId} not found.");

        order.Confirm();
        await orderRepo.SaveAsync(order, ct);
        return true;
    }
}
```

### ২.৩ মডুলার মনোলিথ আর্কিটেকচার (Modular Monolith Architecture)

মডুলার মনোলিথ একটি একক ডিপ্লয়েবল প্রসেসে চলে, কিন্তু অভ্যন্তরীণভাবে ব্যবসায়িক সক্ষমতা বা Bounded Context-এর ভিত্তিতে কঠোরভাবে পৃথক মডিউলে বিভক্ত থাকে যা ইন-মেমোরি ইভেন্ট বা পাবলিক ইন্টারফেসে যোগাযোগ করে।

```
┌─────────────────────────────────────────────────────────────┐
│                    Modular Monolith Host                    │
│                                                             │
│   ┌──────────────────────┐      ┌──────────────────────┐    │
│   │    Orders Module     │      │   Inventory Module   │    │
│   │                      │      │                      │    │
│   │  • Domain Logic      │      │  • Domain Logic      │    │
│   │  • Dedicated Tables  │      │  • Dedicated Tables  │    │
│   └──────────┬───────────┘      └──────────▲───────────┘    │
│              │                             │                │
│              └──── In-Memory Event Bus ────┘                │
│                    (Decoupled Contracts)                    │
└─────────────────────────────────────────────────────────────┘
```

#### সুবিধাসমূহ (Benefits)
- **শূন্য নেটওয়ার্ক ল্যাগ**: ইন-প্রসেস মেথড কল মাইক্রোসেকেন্ডে সম্পন্ন হয়, কোনো নেটওয়ার্ক লেটেন্সি নেই।
- **সহজ রিফ্যাক্টরিং**: আর্কিটেকচার বাউন্ডারি ভুল হলে একটিমাত্র কোডবেসে সাধারণ মেথড রিফ্যাক্টরিং করলেই হয়।
- **অটোমেটেড আর্কিটেকচারাল টেস্ট**: CI পাইপলাইনে কোডের বাউন্ডারি ভায়োলেশন ইউনিট টেস্টের মাধ্যমে নিশ্চিত করা যায় (`NetArchTest.Rules`)।

#### অসুবিধা ও ট্রেড-অফ (Drawbacks & Trade-offs)
- **উচ্চ কোডিং শৃঙ্খলা প্রয়োজন**: কঠোর আর্কিটেকচার টেস্ট না থাকলে ডেভেলপাররা সরাসরি অন্য মডিউলের ইন্টারনাল ক্লাস ব্যবহার করে ফেলতে পারে।
- **শেয়ার্ড প্রসেস রিস্ক**: একটি মডিউলে মেমোরি লিক হলে সম্পূর্ণ প্রসেস ক্র্যাশ করতে পারে।

#### টার্গেট টেকনোলজি ও .NET / C# NuGet প্যাকেজ
- `NetArchTest.Rules`: আর্কিটেকচার ফিটনেস ফাংশন টেস্ট করার জন্য।
- `MassTransit.Mediator`: হাই-থ্রুপুট ইন-মেমোরি মেসেজ বাসের জন্য।

#### মিনিমাল প্রোডাকশন C# কোড উদাহরণ

### ২.৪ মাইক্রোকার্নেল বা প্লাগইন আর্কিটেকচার (Microkernel / Plugin Architecture)

মাইক্রোকার্নেল আর্কিটেকচার একটি ন্যূনতম **কোর সিস্টেম** এবং রানটাইমে লোড হওয়া স্বাধীন **প্লাগইন কম্পোনেন্ট** নিয়ে গঠিত, যা রিকম্পাইলেশন ছাড়াই সিস্টেমকে সম্প্রসারণ করতে দেয়।

```
┌────────────────────────────────────────────────────────┐
│                   Core Microkernel                     │
│                                                        │
│  • Lifecycle Management  • Plugin Registry             │
│  • Extension Point Contracts (IPlugin / IProcessor)    │
└───▲────────────────────▲────────────────────▲──────────┘
    │                    │                    │
┌───┴──────────┐   ┌─────┴────────┐   ┌───────┴──────────┐
│  Stripe      │   │  PayPal      │   │  Crypto Payment  │
│  Payment     │   │  Payment     │   │  Plugin          │
│  Plugin      │   │  Plugin      │   │  (External .dll) │
└──────────────┘   └──────────────┘   └──────────────────┘
```

#### সুবিধাসমূহ (Benefits)
- **উচ্চ সম্প্রসারণযোগ্যতা**: মূল কোডে কোনো হাত না দিয়েই নতুন পেমেন্ট বা ফাইল প্রসেসর প্লাগইন হিসেবে যুক্ত করা যায়।
- **মাল্টি-টেন্যান্ট কাস্টমাইজেশন**: ভিন্ন গ্রাহকের জন্য ভিন্ন প্লাগইন সক্রিয় করা যায়।

#### অসুবিধা ও ট্রেড-অফ (Drawbacks & Trade-offs)
- **প্লাগইন কন্ট্রাক্ট বিবর্তন**: কোর ইন্টারফেস পরিবর্তন হলে পুরনো প্লাগইনগুলো অকেজো হয়ে পড়তে পারে।
- **সিকিউরিটি ও রিসোর্স ঝুঁকি**: খারাপভাবে লিখিত প্লাগইন অতিরিক্ত মেমোরি বা থ্রেড দখল করতে পারে।

#### টার্গেট টেকনোলজি ও .NET / C# NuGet প্যাকেজ
- `System.Runtime.Loader.AssemblyLoadContext`: রানটাইমে ডাইনামিকালি অ্যাসেম্বলি লোড এবং আনলোড করার অফিসিয়াল .NET মেকানিজম।

#### মিনিমাল প্রোডাকশন C# কোড উদাহরণ
```csharp
// Microkernel Architecture: Dynamic Plugin Engine using AssemblyLoadContext
using System.Reflection;
using System.Runtime.Loader;

public interface IPaymentPlugin
{
    string ProviderName { get; }
    Task<bool> ProcessPaymentAsync(decimal amount, CancellationToken ct);
}

public class PluginRegistry
{
    private readonly Dictionary<string, IPaymentPlugin> _plugins = new();

    public void RegisterPlugin(IPaymentPlugin plugin) =>
        _plugins[plugin.ProviderName.ToLowerInvariant()] = plugin;

    public IPaymentPlugin GetPlugin(string providerName) =>
        _plugins.TryGetValue(providerName.ToLowerInvariant(), out var plugin)
            ? plugin
            : throw new NotSupportedException($"Provider '{providerName}' is not installed.");

    public void LoadPluginsFromDirectory(string pluginDirectory)
    {
        foreach (var dllPath in Directory.GetFiles(pluginDirectory, "*.dll"))
        {
            var loadContext = new AssemblyLoadContext(Path.GetFileName(dllPath), isCollectible: true);
            var assembly = loadContext.LoadFromAssemblyPath(Path.GetFullPath(dllPath));

            var pluginTypes = assembly.GetTypes()
                .Where(t => typeof(IPaymentPlugin).IsAssignableFrom(t) && !t.IsInterface && !t.IsAbstract);

            foreach (var type in pluginTypes)
            {
                if (Activator.CreateInstance(type) is IPaymentPlugin pluginInstance)
                    RegisterPlugin(pluginInstance);
            }
        }
    }
}
```

### ২.৫ ইউআই এবং প্রেজেন্টেশন আর্কিটেকচার প্যাটার্নস (MVC, MVP, MVVM, MVI/Flux, BFF)

ইউআই (UI) আর্কিটেকচার প্যাটার্নস মূলত নির্ধারণ করে কীভাবে অ্যাপ্লিকেশনের **Presentation Layer** ইউজার ইন্টারফেস, ভিউ স্টেট (View State) এবং ডোমেইন লজিককে একে অপর থেকে পৃথক রাখবে। সার্ভার-রেন্ডার করা ওয়েব পেজ থেকে শুরু করে আধুনিক রিঅ্যাক্টিভ সিঙ্গেল-পেজ অ্যাপ্লিকেশন (SPA) এবং ক্রস-প্ল্যাটফর্ম মোবাইল অ্যাপে রূপান্তরের সাথে সাথে ইউআই প্যাটার্নগুলোর বিবর্তন ঘটেছে যাতে ম্যানুয়াল স্টেট সিঙ্ক্রোনাইজেশনের ঝামেলা ও অপ্রত্যাশিত বাগ দূর করা যায়।

![UI Architecture Patterns Evolution](images/ui-architecture-patterns.svg)

#### ১. প্রধান ইউআই প্যাটার্নসমূহের তুলনামূলক বিশ্লেষণ

| প্যাটার্ন | কন্ট্রোল বা ডেটা ফ্লো | ভিউ সচেতনতা (View Awareness) | স্টেট সিঙ্ক্রোনাইজেশন | আধুনিক প্রধান ব্যবহার |
| :--- | :--- | :--- | :--- | :--- |
| **MVC** (Model-View-Controller) | User $\rightarrow$ Controller $\rightarrow$ Model $\rightarrow$ View | Controller ভিউ নির্বাচন করে; View মডেল কুয়েরি করে | সার্ভার রিকোয়েস্ট-রেসপন্স চক্র | সার্ভার-সাইড রেন্ডারড (SSR) পোর্টাল (ASP.NET Core MVC, Rails) |
| **MVP** (Model-View-Presenter) | View $\rightleftarrows$ Presenter $\rightleftarrows$ Model | Presenter সরাসরি `IView` ইন্টারফেস রেফারেন্স রাখে | ম্যানুয়াল কোডের মাধ্যমে পরিবর্তন | WinForms, ডেস্কটপ লিগ্যাসি, শতভাগ মকযোগ্য Android MVP |
| **MVVM** (Model-View-ViewModel) | View $\rightleftarrows$ Data Binding $\rightleftarrows$ ViewModel $\rightarrow$ Model | View ভিউমডেলে বাইন্ড করে; ViewModel ভিউ চেনে না | ডিক্লারেটিভ টু-ওয়ে বা ওয়ান-ওয়ে ডেটা বাইন্ডিং | .NET MAUI, WPF, Blazor, Angular, Vue |
| **MVI / Flux** (Unidirectional Flow) | View $\rightarrow$ Intent $\rightarrow$ Reducer $\rightarrow$ State $\rightarrow$ View | View কেবল সিঙ্গেল ইমিউটেবল স্টেট পর্যবেক্ষণ করে | ইউনিডিরেকশনাল ডেটা ফ্লো (UDF) চক্র | Blazor + Fluxor, React/Redux, Jetpack Compose, SwiftUI |
| **BFF** (Backend for Frontend) | Client UI $\rightarrow$ Dedicated BFF Service $\rightarrow$ APIs | নির্দিষ্ট ক্লায়েন্ট প্ল্যাটফর্মের সাথে ১:১ সংযুক্ত | নেটওয়ার্ক এপিআই অ্যাগ্রিগেশন ও ফিল্টারিং | মোবাইল, ওয়েব ও আইওটির জন্য পৃথক অপ্টিমাইজড আর্কিটেকচার |

#### ২. ব্যাকএন্ড-ফর-ফ্রন্টএন্ড (BFF) প্যাটার্ন
মোবাইল অ্যাপ, ডেস্কটপ ব্রাউজার এবং স্মার্টওয়াচের মতো ভিন্ন ভিন্ন ফ্রন্টএন্ডের জন্য একটিমাত্র সাধারণ এপিআই গেটওয়ে ব্যবহার না করে, **BFF প্যাটার্ন** প্রতিটি নির্দিষ্ট ক্লায়েন্টের জন্য একটি পৃথক লাইটওয়েট ব্যাকএন্ড সার্ভিস প্রদান করে:
- **Mobile BFF**: সেলুলার নেটওয়ার্কের ডেটা ও ব্যাটারি বাঁচাতে ৫টি মাইক্রোসার্ভিসের রেসপন্স একত্রিত (aggregate) ও কম্প্রেস করে একটি একক রেসপন্স পাঠায়।
- **Web BFF**: উচ্চগতির ব্রডব্যান্ড ও বড় স্ক্রিনের জন্য রিচ পে-লোড এবং কুকি-ভিত্তিক সেশন পরিচালনা করে।

#### সুবিধাসমূহ (Benefits)
- **কোড-বিহাইন্ড মুক্ত ও উচ্চ টেস্টেবিলিটি**: MVVM এবং MVI-তে সমস্ত ইউআই লজিক সাধারণ C# ক্লাসের ভেতরে থাকে যা কোনো উইন্ডো বা ব্রাউজার রান না করেই ইউনিট টেস্ট করা যায়।
- **পূর্বাভাসযোগ্য স্টেট পরিবর্তন (MVI/Flux)**: স্টেট অপরিবর্তনীয় (immutable) হওয়ায় এবং পিওর রিডিউসার (Reducer) ফাংশনের মাধ্যমে পরিবর্তিত হওয়ায় কোনো রেস কন্ডিশন বা অসঙ্গতি তৈরি হয় না।
- **নেটওয়ার্কের সর্বোচ্চ দক্ষতা (BFF)**: মোবাইল ডিভাইসের জন্য অতিরিক্ত ডেটা পাঠানো (Over-fetching) বা বারবার কল করা (Under-fetching) রোধ করে।

#### অসুবিধা ও ট্রেড-অফ (Drawbacks & Trade-offs)
- **টু-ওয়ে বাইন্ডিংয়ের জটিলতা (MVVM)**: জটিল ফর্মে টু-ওয়ে ডেটা বাইন্ডিং অসাবধানতাবশত ক্যাসকেডিং প্রোপার্টি আপডেটের লুপ এবং পারফরম্যান্স সমস্যা তৈরি করতে পারে।
- **বয়লারপ্লেট কোড (MVI/Flux)**: প্রতিটি সাধারণ ইন্টারঅ্যাকশনের জন্য একটি Action, Reducer, State রেকর্ড এবং Effect তৈরি করতে হয়।
- **অপারেশনাল ওভারহেড (BFF)**: প্রতিটি ফ্রন্টএন্ডের জন্য আলাদা BFF সার্ভিস বজায় রাখতে ডিপ্লয়মেন্ট পাইপলাইনের সংখ্যা বৃদ্ধি পায়।

#### টার্গেট টেকনোলজি ও .NET / C# NuGet প্যাকেজ
- `CommunityToolkit.Mvvm`: .NET MAUI, WPF এবং Blazor-এর জন্য সোর্স-জেনারেটর ভিত্তিক হাই-পারফরম্যান্স MVVM লাইব্রেরি।
- `Fluxor.Blazor.Web`: Blazor WebAssembly ও Server-এর জন্য Redux/MVI স্টেট ম্যানেজমেন্ট ফ্রেমওয়ার্ক।
- `Yarp.ReverseProxy`: ASP.NET Core-এ হাই-পারফরম্যান্স BFF গেটওয়ে তৈরির জন্য আদর্শ।

### ৩.১ ইভেন্ট-ড্রিভেন প্যাটার্নস: ব্রোকার বনাম মিডিয়েটর টপোলজি

ইভেন্ট-ড্রিভেন আর্কিটেকচারে (EDA) স্বাধীন উপাদানগুলো অপরিবর্তনীয় **Events** উৎপাদন এবং গ্রহণের মাধ্যমে যোগাযোগ করে।

![EDA Broker vs Mediator](images/eda-broker-vs-mediator.svg)

- **Broker Topology (Choreography)**: কোনো কেন্দ্রীয় সমন্বয়কারী থাকে না। সার্ভিসগুলো ব্রোকারে (Kafka/RabbitMQ) ইভেন্ট পোস্ট করে এবং অন্যরা স্বয়ংক্রিয়ভাবে কাজ করে। অতি উচ্চ থ্রুপুট নিশ্চিত করে।
- **Mediator Topology (Orchestration)**: একটি কেন্দ্রীয় অর্কেস্ট্রেটর (Event Mediator) ব্যবসায়িক প্রক্রিয়ার বিভিন্ন ধাপ নিয়ন্ত্রণ করে এবং ব্যর্থ হলে কম্পেনসেশন পরিচালনা করে (Saga Pattern)।

#### সুবিধাসমূহ (Benefits)
- **টেম্পোরাল ডিকাপলিং**: প্রডিউসার এবং কনজিউমারকে একই সময়ে অনলাইনে থাকার প্রয়োজন নেই।
- **ব্রডকাস্ট স্কেলেবিলিটি**: একই ইভেন্ট একাধিক স্বাধীন কনজিউমার গ্রুপের কাছে ফ্যান-আউট করা যায়।

#### অসুবিধা ও ট্রেড-অফ (Drawbacks & Trade-offs)
- **ট্রেসেবিলিটি চ্যালেঞ্জ**: ডিস্ট্রিবিউটেড মেসেজ কিউতে এন্ড-টু-এন্ড ফ্লো ট্র্যাক করা জটিল।
- **অ্যাট-লিস্ট-ওয়ান্স ডেলিভারি**: ডুপ্লিকেট মেসেজ ঠেকাতে কনজিউমারদের কঠোরভাবে আইডেমপোটেন্ট (Idempotent) হতে হয়।

#### টার্গেট টেকনোলজি ও .NET / C# NuGet প্যাকেজ
- `MassTransit`: RabbitMQ, Kafka, Azure Service Bus-এর উপর এন্টারপ্রাইজ সার্ভিস বাস অ্যাবস্ট্রাকশন।

#### মিনিমাল প্রোডাকশন C# কোড উদাহরণ
```csharp
// MassTransit Consumer with Idempotency & Chained Event Emission
using MassTransit;
using Microsoft.Extensions.Logging;

public record OrderSubmittedEvent(Guid OrderId, decimal Amount, DateTime CreatedAt);
public record InvoiceGeneratedEvent(Guid OrderId, decimal Amount);

public class OrderSubmittedConsumer(ILogger<OrderSubmittedConsumer> logger) 
    : IConsumer<OrderSubmittedEvent>
{
    public async Task Consume(ConsumeContext<OrderSubmittedEvent> context)
    {
        var message = context.Message;
        logger.LogInformation("Processing OrderSubmittedEvent: {OrderId}", message.OrderId);

        var messageId = context.MessageId ?? throw new InvalidOperationException("Missing MessageId");
        
        await Task.Delay(50, context.CancellationToken); // বিজনেস লজিক

        // একই কনটেক্সটে পরবর্তী ইভেন্ট পাঠানো
        await context.Publish(new InvoiceGeneratedEvent(message.OrderId, message.Amount));
    }
}
```

### ৩.২ সিকিউআরএস এবং ইভেন্ট সোর্সিং (CQRS & Event Sourcing)

**CQRS** রাইট (Commands) এবং রিডকে (Queries) সম্পূর্ণ পৃথক করে। **ইভেন্ট সোর্সিং** বর্তমান স্টেট ওভাররাইট করার বদলে স্টেট পরিবর্তনের অপরিবর্তনীয় ডোমেইন ইভেন্ট লগ হিসেবে সংরক্ষণ করে।

![CQRS & Event Sourcing Flow](images/cqrs-event-sourcing-flow.svg)

সুবিধাসমূহ (Benefits):
- রিড এবং রাইট কাজের চাপ উল্লেখযোগ্যভাবে আলাদা হলে
- জটিল বা বিপুল সংখ্যক রিড কুয়েরি থাকলে
- বিভিন্ন ইউআই স্ক্রিনে ভিন্ন ভিন্ন ডেটা প্রজেকশন প্রয়োজন হলে
- ব্যবসায়িক ডোমেইনে জটিল ইনভেরিয়েন্ট বা ব্যবসায়িক নিয়ম থাকলে
- রিড এবং রাইট সাইডকে স্বাধীনভাবে স্কেল করতে হলে
- একাধিক বিশেষায়িত রিড মডেলের প্রয়োজন হলে
- সিস্টেমটি ইতিমধ্যে ইভেন্ট বা অ্যাসিঙ্ক্রোনাস প্রসেসিং ব্যবহার করলে

কখন সম্পূর্ণ CQRS এড়িয়ে চলবেন:
- আপনার অ্যাপ্লিকেশনটি প্রধানত সাধারণ CRUD প্রকৃতির হলে।
- রিড এবং রাইট ট্রাফিক সীমিত হলে।
- একটি একক ডেটাবেস এবং একটি একক মডেল দিয়েই সন্তোষজনক পারফরম্যান্স পাওয়া গেলে।
- অতিরিক্ত অপারেশনাল জটিলতা সামলানোর মতো রিসোর্স বা সময় টিমের না থাকলে।

ট্রেড-অফ ও চ্যালেঞ্জ (Trade-offs):
- রক্ষণাবেক্ষণের জন্য হ্যান্ডলার, ইন্টারফেস এবং কোডের পরিমাণ অনেক বৃদ্ধি পায়
- অতিরিক্ত রিড-মডেল ডিজাইন এবং ডেটা সিনক্রোনাইজেশন ওভারহেড
- আলাদাভাবে ডিপ্লয় করলে অবকাঠামোগত খরচ বৃদ্ধি পায়
- ডোমেইন নিয়মের পুনরাবৃত্তি রোধে কঠোর শৃঙ্খলার প্রয়োজন
- ডেটা সিনক্রোনাইজেশন এবং ইভেনচুয়াল কনসিস্টেন্সি
- ডেটার ডুপ্লিকেশন
- একাধিক ডিপ্লয়মেন্ট উপাদান এবং বর্ধিত মনিটরিং
- রিড মডেল পুনরায় তৈরি (Rebuild) বা মেরামত করা
- ডিবাগিং তুলনামূলকভাবে জটিল

#### সুবিধাসমূহ (Benefits)
- **নিখুঁত অডিট ট্রেইল**: সিস্টেমের প্রতিটি পরিবর্তন কার দ্বারা ও কখন সম্পন্ন হয়েছে তা অপরিবর্তনীয়ভাবে সংরক্ষিত থাকে।
- **টাইম ট্রাভেল ও প্রজেকশন**: পুরনো ইভেন্ট লগ থেকে সম্পূর্ণ নতুন কোনো রিড মডেল বা অ্যানালিটিক্স ভিউ তৈরি করা যায়।
- **স্বাধীন রিড/রাইট অপ্টিমাইজেশন**: রিড সাইডে Elasticsearch/Redis ব্যবহার করা যায়, আর রাইট সাইডে কেবল সিকোয়েন্সিয়াল অ্যাপেন্ড ঘটে।

#### অসুবিধা ও ট্রেড-অফ (Drawbacks & Trade-offs)
- **ইভেনচুয়াল কনসিস্টেন্সি ল্যাগ**: ইভেন্ট রাইট হওয়া এবং রিড প্রজেকশন আপডেট হওয়ার মাঝে সামান্য সময়ের ব্যবধান থাকে।
- **ইভেন্ট স্কিমা বিবর্তন**: সময়ের সাথে সাথে ইভেন্ট স্কিমা পরিবর্তিত হলে আপকাস্টিং নিশ্চিত করতে হয়।

#### ইভেন্ট সোর্সিং (Event Sourcing):
ইভেন্ট সোর্সিং হলো একটি অ্যাপ্লিকেশন-লেভেল আর্কিটেকচারাল প্যাটার্ন যেখানে কোনো এন্টিটির কেবল সর্বশেষ স্টেট সংরক্ষণ করার পরিবর্তে প্রতিটি স্টেট পরিবর্তনকারী ঘটনাকে একটি অপরিবর্তনীয় অ্যাপেন্ড-অনলি রেকর্ড হিসেবে সংরক্ষণ করা হয়।

#### টার্গেট টেকনোলজি ও .NET / C# NuGet প্যাকেজ
- `Marten`: PostgreSQL-এর উপর নির্মিত .NET ট্রানজ্যাকশনাল ডকুমেন্ট ডিবি এবং ইভেন্ট স্টোর।

#### মিনিমাল প্রোডাকশন C# কোড উদাহরণ
```csharp
// Event Sourced Aggregate using C# 12 & Marten Event Store pattern
public record BankAccountCreated(Guid AccountId, string Owner, decimal InitialBalance);
public record MoneyDeposited(Guid AccountId, decimal Amount);
public record MoneyWithdrawn(Guid AccountId, decimal Amount);

public class BankAccountAggregate
{
    public Guid Id { get; private set; }
    public decimal Balance { get; private set; }

    public void Apply(BankAccountCreated @event)
    {
        Id = @event.AccountId;
        Balance = @event.InitialBalance;
    }

    public void Apply(MoneyDeposited @event) => Balance += @event.Amount;
    public void Apply(MoneyWithdrawn @event) => Balance -= @event.Amount;

    public MoneyWithdrawn Withdraw(decimal amount)
    {
        if (amount <= 0) throw new ArgumentException("Withdrawal amount must be positive.");
        if (Balance - amount < 0) throw new InvalidOperationException("Insufficient funds.");
        return new MoneyWithdrawn(Id, amount);
    }
}

public record AccountSummaryView(Guid Id, string Owner, decimal Balance, DateTime LastUpdatedUtc);

public class AccountSummaryProjection
{
    public AccountSummaryView Create(BankAccountCreated created) =>
        new(created.AccountId, created.Owner, created.InitialBalance, DateTime.UtcNow);

    public AccountSummaryView Apply(MoneyDeposited deposited, AccountSummaryView current) =>
        current with { Balance = current.Balance + deposited.Amount, LastUpdatedUtc = DateTime.UtcNow };
}
```

### ৩.৩ ট্রানজ্যাকশনাল আউটবক্স প্যাটার্ন (Transactional Outbox Pattern)

যখন কোনো অ্যাপ্লিকেশনকে ডেটাবেস আপডেট এবং মেসেজ ব্রোকারে ইভেন্ট পাঠানো—উভয় কাজ করতে হয়, তখন আলাদাভাবে কল করলে **Dual-Write Problem** তৈরি হয়: ডেটাবেসে কমিট হওয়ার পর যদি ব্রোকার কল নেটওয়ার্ক ফেইলিউরে ব্যর্থ হয়, তবে ডাউনস্ট্রিম সার্ভিস কখনোই তা জানতে পারে না।

```
┌────────────────────────────────────────────────────────┐
│               Single Local Transaction                 │
│                                                        │
│  1. INSERT INTO Orders (...)                           │
│  2. INSERT INTO OutboxMessages (EventPayload, Status)  │
└──────────────────────────┬─────────────────────────────┘
                           │ (Atomic Commit)
                           ▼
                  Relational Database
                           │
             ┌─────────────┴─────────────┐
             ▼                           ▼
    Background Outbox Worker        Debezium CDC
             │                           │
             └─────────────┬─────────────┘
                           ▼
               Message Broker (RabbitMQ)
```

#### সুবিধাসমূহ (Benefits)
- **শতভাগ ডেটা সামঞ্জস্য (Zero Inconsistency)**: ডেটাবেস ট্রানজ্যাকশন সফল হলে ইভেন্ট নিশ্চিতভাবে ব্রোকারে পৌঁছাবে।
- **অ্যাট-লিস্ট-ওয়ান্স ডেলিভারি**: ব্রোকার সাময়িক ডাউন থাকলেও নেটওয়ার্ক ঠিক হলে মেসেজ নিশ্চিতভাবে ডেলিভার হয়।

#### অসুবিধা ও ট্রেড-অফ (Drawbacks & Trade-offs)
- **টেবিল পোলিং ওভারহেড**: আউটবক্স টেবিলে ব্যাকগ্রাউন্ডে পোলিং বা সিডিসি (Debezium) টুলের প্রয়োজন হয়।

#### টার্গেট টেকনোলজি ও .NET / C# NuGet প্যাকেজ
- `MassTransit.EntityFrameworkCore`: EF Core-এর সাথে বিল্ট-ইন আউটবক্স বাস্তবায়ন।

### ৩.৪ ডেটা এক্সেস এবং পারসিস্টেন্স প্যাটার্নস: রিপোজিটরি এবং ইউনিট অফ ওয়ার্ক (Repository & Unit of Work)

**রিপোজিটরি প্যাটার্ন (Repository Pattern)** এবং **ইউনিট অফ ওয়ার্ক প্যাটার্ন (Unit of Work Pattern)** (মার্টিন ফাউলারের *Patterns of Enterprise Application Architecture* এবং এরিক ইভান্সের *Domain-Driven Design* গ্রন্থে সংজ্ঞায়িত) মূলত **লেভেল ৩: ট্যাকটিক্যাল আর্কিটেকচারাল এবং ডিজাইন প্যাটার্নস**-এর অন্তর্ভুক্ত। এগুলো **লেভেল ২ অ্যাপ্লিকেশন আর্কিটেকচার**-এর অভ্যন্তরে ডেটা পারসিস্টেন্স নিশ্চিত করার মূল ভিত্তি হিসেবে কাজ করে:
- **লেয়ারড (N-Tier) আর্কিটেকচারে**: এগুলো **Data Access Layer (DAL)** হিসেবে অবস্থান করে।
- **ক্লিন বা হেক্সাগোনাল আর্কিটেকচারে**: রিপোজিটরি এবং ইউনিট অফ ওয়ার্কের ইন্টারফেসগুলো (`IOrderRepository`, `IUnitOfWork`) অ্যাপ্লিকেশনের ভেতরে **Outbound Ports** হিসেবে সংজ্ঞায়িত হয় এবং তাদের ডাটাবেস বাস্তবায়ন (`EfOrderRepository`, `EfUnitOfWork`) ইনফ্রাস্ট্রাকচার স্তরে **Secondary Adapters** হিসেবে কাজ করে।

```
┌─────────────────────────────────────────────────────────────┐
│                 Application Layer (Use Case)                │
│                                                             │
│  var order = await _orderRepo.GetByIdAsync(orderId);       │
│  order.AddDiscount(couponCode);                             │
│  await _orderRepo.UpdateAsync(order);                       │
│  await _unitOfWork.SaveChangesAsync(); // অ্যাটমিক কমিট     │
└──────────────┬───────────────────────────────▲──────────────┘
               │ (Calls Ports)                 │
               ▼                               ▼
┌──────────────────────────────┐ ┌─────────────────────────────┐
│      Repository Pattern      │ │    Unit of Work Pattern     │
│   (Collection Abstraction)   │ │  (Transaction Coordinator)  │
│  • GetByIdAsync()            │ │  • পরিবর্তন ট্র্যাক করে     │
│  • AddAsync()                │ │  • রোলব্যাক নিয়ন্ত্রণ করে  │
│  • Remove()                  │ │  • SaveChangesAsync()       │
└──────────────┬───────────────┘ └─────────────┬───────────────┘
               │                               │
               └───────────────┬───────────────┘
                               ▼
┌─────────────────────────────────────────────────────────────┐
│    Infrastructure / EF Core DbContext (Outbound Adapter)    │
│  • DbContext = Unit of Work   • DbSet<T> = Repository       │
└──────────────────────────────┬──────────────────────────────┘
                               ▼
                      Relational Database
```

#### রিপোজিটরি প্যাটার্নের মেকানিজম ও উদ্দেশ্য
ডোমেইন এবং ডেটাবেস ম্যাপিং লেয়ারের মধ্যে একটি ইন-মেমোরি কালেকশনের মতো কাজ করে। এটি কুয়েরি লজিক, ক্যাশিং এবং রিলেশনাল রো-কে রিচ ডোমেইন মডেলে রূপান্তর করার সম্পূর্ণ দায়িত্ব নিজের ভেতর আবদ্ধ (encapsulate) রাখে।

#### ইউনিট অফ ওয়ার্ক প্যাটার্নের মেকানিজম ও উদ্দেশ্য
একটি নির্দিষ্ট ব্যবসায়িক ট্রানজ্যাকশনে একাধিক রিপোজিটরির মাধ্যমে ডেটার যে পরিবর্তন ঘটে তা ট্র্যাক করে এবং সবশেষে একটি একক ডাটাবেস ট্রানজ্যাকশনে অ্যাটমিকভাবে কমিট (সব সফল অথবা সব বাতিল) নিশ্চিত করে।

#### সুবিধাসমূহ (Benefits)
- **ডোমেইন ও এসকিউএল লজিকের সম্পূর্ণ বিচ্ছেদ**: ব্যবসায়িক ইউজ কেস ডেটাবেস ও এসকিউএল কোড থেকে সম্পূর্ণ স্বাধীন থাকে এবং পিওর ডোমেইন টার্মে কথা বলে (`_repo.GetPendingOrdersForCustomer(id)`)।
- **উচ্চ টেস্টেবিলিটি ও মকিং সুবিধা**: আসল ডেটাবেস চালু না করেই ইউনিট টেস্টে রিপোজিটরি ইন্টারফেস মক করে সমস্ত ইউজ কেস পরীক্ষা করা যায়।
- **অ্যাটমিক ট্রানজ্যাকশন অখণ্ডতা**: ইউনিট অফ ওয়ার্ক নিশ্চিত করে যে একাধিক রিপোজিটরিতে ডেটা আপডেট হলেও তা যেন একটিমাত্র ডেটাবেস ট্রানজ্যাকশনে সংরক্ষিত হয়।

#### অসুবিধা ও ট্রেড-অফ (Drawbacks & Trade-offs)
- **EF Core এবং "Leaky Abstraction" বিতর্ক**: আধুনিক .NET-এ Entity Framework Core-এর `DbContext` মূলত **ইতিমধ্যেই একটি Unit of Work** এবং `DbSet<T>` মূলত **ইতিমধ্যেই একটি Repository**।
  - *অ্যান্টি-প্যাটার্ন*: সাধারণ জেনেরিক `IRepository<T>` তৈরি করে তার মাধ্যমে `IQueryable<T>` বাইরে এক্সপোজ করা একটি গুরুতর অ্যান্টি-প্যাটার্ন, কারণ এতে ডেটাবেস কুয়েরি লজিক সার্ভিস লেয়ারে লিক হয়ে যায়।
  - *বেস্ট প্র্যাকটিস*: EF Core-এর উপর রিপোজিটরি ব্যবহার করতে চাইলে সর্বদা **Domain-Specific Repository** (যেমন `GetOrderWithItemsAsync`) তৈরি করা উচিত, অপ্রয়োজনীয় জেনেরিক র‍্যাপার নয়।

#### টার্গেট টেকনোলজি ও .NET / C# NuGet প্যাকেজ
- `Microsoft.EntityFrameworkCore`: .NET-এর নেটিভ ওআরএম যা বিল্ট-ইন Unit of Work (`DbContext`) ও Repository (`DbSet<T>`) প্রদান করে।
- `Ardalis.Specification`: রিপোজিটরির কুয়েরি লজিক টাইপ-সেফ ও ক্লিন রাখার জন্য স্পেসিফিকেশন প্যাটার্ন লাইব্রেরি।

#### মিনিমাল প্রোডাকশন C# কোড উদাহরণ
```csharp
// Production .NET 9: Domain-Specific Repository & Unit of Work with EF Core
using Microsoft.EntityFrameworkCore;

// ১. Outbound Port ইন্টারফেস (Application Layer-এ সংজ্ঞায়িত)
public interface IOrderRepository
{
    Task<Order?> GetByIdWithItemsAsync(Guid id, CancellationToken ct = default);
    Task AddAsync(Order order, CancellationToken ct = default);
}

public interface IUnitOfWork
{
    Task<int> SaveChangesAsync(CancellationToken ct = default);
}

// ২. Outbound Adapters (EF Core ব্যবহার করে Infrastructure Layer-এ বাস্তবায়িত)
public class AppDbContext(DbContextOptions<AppDbContext> options) : DbContext(options), IUnitOfWork
{
    public DbSet<Order> Orders => Set<Order>();
    // DbContext.SaveChangesAsync() স্বয়ংক্রিয়ভাবে IUnitOfWork.SaveChangesAsync() পূর্ণ করে
}

public class OrderRepository(AppDbContext db) : IOrderRepository
{
    public async Task<Order?> GetByIdWithItemsAsync(Guid id, CancellationToken ct = default)
    {
        // EF-নির্দিষ্ট Eager Loading কেবল ইনফ্রাস্ট্রাকচার অ্যাডাপ্টারের ভেতরে সীমাবদ্ধ থাকে
        return await db.Orders
            .Include("_items")
            .FirstOrDefaultAsync(o => o.Id == id, ct);
    }

    public async Task AddAsync(Order order, CancellationToken ct = default)
    {
        await db.Orders.AddAsync(order, ct);
    }
}

// ৩. Application Use Case: রিপোজিটরি ও ইউনিট অফ ওয়ার্কের মাধ্যমে অ্যাটমিক অপারেশন
public class CancelOrderHandler(IOrderRepository orderRepo, IUnitOfWork unitOfWork)
{
    public async Task<bool> Handle(Guid orderId, CancellationToken ct)
    {
        var order = await orderRepo.GetByIdWithItemsAsync(orderId, ct)
            ?? throw new KeyNotFoundException($"Order {orderId} not found.");

        order.Confirm(); // ডোমেইন স্টেট পরিবর্তন

        // ইউনিট অফ ওয়ার্ক সমস্ত পরিবর্তন অ্যাটমিকভাবে ডেটাবেসে কমিট করে
        await unitOfWork.SaveChangesAsync(ct);
        return true;
    }
}
```

## পর্ব ৫: স্টাফ/প্রিন্সিপাল-লেভেল আর্কিটেকচারাল ইন্টারভিউ প্রশ্নোত্তর (Senior/Staff Interview Q&A)

### প্রশ্ন ১: একটি একক এন্টারপ্রাইজ সলিউশনে সিস্টেম আর্কিটেকচার, অ্যাপ্লিকেশন আর্কিটেকচার এবং আর্কিটেকচারাল প্যাটার্নস কীভাবে পরস্পরের সাথে সম্পর্কিত?
**উত্তর**:
এগুলো মূলত আর্কিটেকচারের তিনটি নেস্টেড স্তর (Nested Levels of Abstraction):
১. **সিস্টেম আর্কিটেকচার (Macro-Level)**: সমগ্র সিস্টেমের ডিপ্লয়মেন্ট টপোলজি, ফিজিক্যাল নেটওয়ার্ক সীমানা এবং আন্তঃ-প্রসেস যোগাযোগ নির্ধারণ করে। উদাহরণস্বরূপ, সম্পূর্ণ এন্টারপ্রাইজটি একটি **Asynchronous Microservices Topology**-তে ডিপ্লয় করা যেখানে পৃথক কন্টেইনারগুলো RabbitMQ-এর মাধ্যমে যোগাযোগ করে এবং সামনে একটি YARP রিভার্স প্রক্সি গেটওয়ে রয়েছে।
২. **অ্যাপ্লিকেশন আর্কিটেকচার (Meso-Level)**: একটি নির্দিষ্ট ডিপ্লয়েবল সার্ভিস প্রসেসের *ভেতরে* সোর্স কোড কীভাবে সাজানো হবে তা নির্ধারণ করে। `OrderService`-এর ভেতরে কোডবেসটি **Clean Architecture (Ports and Adapters)** অনুসরণ করে, যাতে ডোমেইন এন্টিটি ও বিজনেস লজিক কোনো ফ্রেমওয়ার্ক (ASP.NET Core) বা ডেটাবেস লাইব্রেরি (EF Core) দ্বারা দূষিত না হয়।
৩. **আর্কিটেকচারাল প্যাটার্নস (Tactical-Level)**: স্থানীয় ও নির্দিষ্ট কাঠামোগত সমস্যা সমাধানের জন্য সুনির্দিষ্ট মেকানিজম। `OrderService`-এর ভেতরে 2PC ডিস্ট্রিবিউটেড ট্রানজ্যাকশন ছাড়া পার্সিস্টেন্স ও ব্রোকার মেসেজিং সমন্বয় করতে **Transactional Outbox Pattern** এবং দ্রুতগতির কোয়েরি পারফরম্যান্সের জন্য **CQRS** প্যাটার্ন ব্যবহার করা হয়েছে।

### প্রশ্ন ২: লজিক্যাল লেয়ার (Logical Layers) এবং ফিজিক্যাল টিয়ার (Physical Tiers)-এর মধ্যকার মৌলিক পার্থক্য কী, এবং এদের গুলিয়ে ফেললে সিস্টেমে কী বিপর্যয় ঘটে?
**উত্তর**:
- **লেয়ার (Logical Boundaries)**: সোর্স কোডের অনুভূমিক বিভাজন (Presentation, Business Logic, Data Access) যা **একটিমাত্র ওএস প্রসেস এবং মেমোরি স্পেসের ভেতরে** কার্যকর হয়। এদের মধ্যকার যোগাযোগ ইন-মেমোরি মেথড কলের মাধ্যমে শূন্য নেটওয়ার্ক সিরিয়ালাইজেশন খরচে ন্যানোসেকেন্ডে ঘটে।
- **টিয়ার (Physical Boundaries)**: অ্যাপ্লিকেশনটির কার্যকর অংশগুলো শারীরিকভাবে **পৃথক হার্ডওয়্যার সার্ভার, ভার্চুয়াল মেশিন বা নেটওয়ার্ক সাবনেটে** বিভক্ত থাকে, যা সিরিয়ালাইজড নেটওয়ার্ক প্রোটোকলের (HTTPS, gRPC, TCP) মাধ্যমে যোগাযোগ করে।
- **ডিজাইন বিপর্যয়**: লেয়ার এবং টিয়ারকে গুলিয়ে ফেললে ইঞ্জিনিয়াররা অপ্রয়োজনে লজিক্যাল লেয়ারগুলোকে পৃথক ফিজিক্যাল সার্ভারে বসিয়ে দেন (যেমন সার্ভার A-তে সার্ভিস লেয়ার এবং সার্ভার B-তে ডেটা অ্যাক্সেস লেয়ার)। ফলে প্রতিটি মেথড কলে ১০–৫০ মিলিসেকেন্ড নেটওয়ার্ক লেটেন্সি, সিরিয়ালাইজেশন ওভারহেড এবং সংযোগ ত্রুটির ঝুঁকি তৈরি হয়—যা একটি ধীরগতির ভঙ্গুর "Distributed Monolith" তৈরি করে। একটি সুপরিকল্পিত ৩-লেয়ার সিস্টেম ১টি ফিজিক্যাল টিয়ার (মনোলিথ) অথবা ৩টি ফিজিক্যাল টিয়ার (ক্লায়েন্ট $\rightarrow$ DMZ প্রক্সি $\rightarrow$ অ্যাপ সার্ভার $\rightarrow$ ডেটাবেস ক্লাস্টার) হিসেবে সফলভাবে চলতে পারে।

### প্রশ্ন ৩: প্রাথমিক পর্যায়ের স্টার্টআপের জন্য মাইক্রোসার্ভিসের চেয়ে মডুলার মনোলিথ কেন বহুগুণ শ্রেষ্ঠ, এবং কনওয়ের সূত্র (Conway's Law) কীভাবে এটি নিয়ন্ত্রণ করে?
**উত্তর**:
- **ডোমেইন অনিশ্চয়তা**: প্রাথমিক অবস্থায় ব্যবসায়িক বাউন্ডারি প্রায়শই পরিবর্তিত হয়। আগেই মাইক্রোসার্ভিসে বিভক্ত করলে নেটওয়ার্ক সীমানা ভুল জায়গায় নির্ধারিত হয়। পরবর্তীকালে বাউন্ডারি পরিবর্তন করতে গেলে নেটওয়ার্ক API পরিবর্তন, ডেটাবেস স্কিমা মাইগ্রেশন এবং ক্রস-টিম সমন্বয়ের বিশাল ওভারহেড তৈরি হয়।
- **ইন-প্রসেস রিফ্যাক্টরিং**: মডুলার মনোলিথে মডিউলগুলো সি-শার্পের নেমস্পেস ও প্রজেক্ট অ্যাসেম্বলি দ্বারা পৃথক থাকে। ডোমেইন পরিবর্তন হলে কয়েক মিনিটের মধ্যে আইডিই (IDE) রিফ্যাক্টরিং টুলের মাধ্যমে কোড পুনর্বিন্যাস করা যায়।
- **আর্কিটেকচারাল টেস্ট**: সিআই/সিডি পাইপলাইনে `NetArchTest.Rules` ব্যবহার করে মডিউল বাউন্ডারি কঠোরভাবে রক্ষা করা যায় (যেমন: `Ordering.Domain` যেন কখনো `Billing.Infrastructure`-কে রেফারেন্স না করে)।
- **কনওয়ের সূত্র (Conway's Law)**: ৪-৮ জনের একটি ছোট দলের পক্ষে ২০টি মাইক্রোসার্ভিসের স্বতন্ত্র ডিপ্লয়মেন্ট ও অপারেশনাল জটিলতা সামলানো অসম্ভব। প্রতিষ্ঠান যখন একাধিক স্বায়ত্তশাসিত, ক্রস-ফাংশনাল টিমে বিভক্ত হয়, কেবল তখনই মাইক্রোসার্ভিস বাস্তব উপযোগিতা প্রদান করে।

### প্রশ্ন ৪: "Distributed Monolith" অ্যান্টি-প্যাটার্নের মূল কারণ, লক্ষণ এবং এটি সমাধানের উপায় কী?
**উত্তর**:
ডিস্ট্রিবিউটেড মনোলিথ মাইক্রোসার্ভিসের অপারেশনাল জটিলতা ও নেটওয়ার্ক লেটেন্সি ধারণ করে, কিন্তু একই সাথে মনোলিথের টাইট কাপলিং ও ভঙ্গুরতা বহন করে।
- **প্রধান লক্ষণসমূহ**:
  ১. *লকস্টেপ ডিপ্লয়মেন্ট*: সার্ভিস A ডিপ্লয় করতে হলে শেয়ার্ড কন্ট্রাক্টের কারণে সার্ভিস B এবং C-কেও বাধ্যতামূলকভাবে একই সাথে ডিপ্লয় করতে হয়।
  ২. *ক্যাসকেডিং সিঙ্ক্রোনাস HTTP চেইন*: সার্ভিস A কল করে B-কে, B কল করে C-কে এবং C কল করে D-কে। সামগ্রিক লেটেন্সি যোগফল আকারে বাড়ে ($T = t_1 + t_2 + t_3 + t_4$), আর সিস্টেম অ্যাভেইলেবিলিটি গুণিতক হারে হ্রাস পায় ($A = a_1 \times a_2 \times a_3 \times a_4$)।
  ৩. *শেয়ার্ড ডেটাবেস টেবিল*: একাধিক মাইক্রোসার্ভিস সরাসরি একই এসকিউএল টেবিল শেয়ার ও কুয়েরি করে।
- **সমাধানের কৌশল**:
  ১. সিঙ্ক্রোনাস REST কলের পরিবর্তে অ্যাসিঙ্ক্রোনাস ইভেন্ট-চালিত মেসেজিং (RabbitMQ/Kafka) গ্রহণ করা।
  ২. কঠোর **Database-per-Service** নীতি কার্যকর করা; অন্য সার্ভিসের ডেটা কেবল API বা রেপ্লিকেটেড রিড মডেলের মাধ্যমে নেওয়া।
  ৩. পারস্পরিক নির্ভরশীলতা দূর করা না গেলে সার্ভিসগুলোকে পুনরায় একটি একক মডুলার মনোলিথে একত্রিত করা।

### প্রশ্ন ৫: সার্ভিস-ওরিয়েন্টেড আর্কিটেকচার (SOA) কীভাবে মাইক্রোসার্ভিসেস থেকে মৌলিকভাবে আলাদা ("Smart Pipes vs. Dumb Endpoints")?
**উত্তর**:
- **SOA ("স্মার্ট পাইপস, ডাম্ব এন্ডপয়েন্টস")**: আর্কিটেকচারের কেন্দ্রবিন্দু হলো একটি **Enterprise Service Bus (ESB)**। ESB অত্যন্ত বুদ্ধিমান ও ভারী; এতে মেসেজ রূপান্তর (XML থেকে JSON), প্রোটোকল ট্রান্সলেশন (SOAP থেকে REST) এবং জটিল রাউটিং লজিক থাকে। এন্ডপয়েন্টগুলো সাধারণত প্রতিষ্ঠানের বিশাল লিগ্যাসি সিস্টেম (SAP, Mainframe) যা একটি শেয়ার্ড এন্টারপ্রাইজ ডেটাবেস ব্যবহার করে।
- **মাইক্রোসার্ভিসেস ("ডাম্ব পাইপস, স্মার্ট এন্ডপয়েন্টস")**: মেসেজ ব্রোকার (Kafka, RabbitMQ) বা HTTP কোনো বিজনেস লজিক রাখে না—এটি কেবল র বাইট আদান-প্রদান করে। সমস্ত ডোমেইন লজিক, ভ্যালিডেশন ও রূপান্তর স্বাধীন মাইক্রোসার্ভিসের ভেতরে থাকে এবং প্রতিটি সার্ভিস নিজস্ব ডেটাবেসের একক মালিক।
- **ESB-ভিত্তিক SOA ব্যর্থ হওয়ার কারণ**: ESB নিজেই একটি দানবীয়, জটিল ও ধীরগতির মনোলিথে পরিণত হয় এবং এর ওপর বিশেষ ভেন্ডর টিমের একচ্ছত্র নিয়ন্ত্রণ থাকায় দ্রুত রিলিজ দেওয়া অসম্ভব হয়ে পড়ে।
- **আধুনিক .NET-এর ভূমিকা**: লিগ্যাসি ইন্টিগ্রেশনের জন্য আধুনিক .NET 8/9-এ `CoreWCF` (SOAP-এর জন্য) এবং কোনো ভারী সেন্ট্রাল প্রোপাইটরি ESB ছাড়াই মেসেজ আদান-প্রদানের জন্য `MassTransit` ব্যবহৃত হয়।

### প্রশ্ন ৬: ক্লাউড হাইপারস্কেল প্ল্যাটফর্মগুলোতে সেল-বেসড (Cellular) আর্কিটেকচার কীভাবে ব্লাস্ট রেডিয়াস সীমাবদ্ধ করে?
**উত্তর**:
বিশ্বের শীর্ষস্থানীয় হাইপারস্কেল ক্লাউড প্ল্যাটফর্মগুলো (যেমন: AWS Route 53, Slack, Netflix, Stripe) একক বিশাল ক্লাস্টারের সীমাবদ্ধতা কাটিয়ে উঠতে সেল-বেসড আর্কিটেকচার ব্যবহার করে:
- **সেলের গঠন (Cell Anatomy)**: সম্পূর্ণ সিস্টেমটি (অ্যাপ্লিকেশন কম্পিউট, ডেটাবেস, ক্যাশ, মেসেজ কিউ) কতগুলো হুবহু একই রকম, ক্ষুদ্র ও সম্পূর্ণ স্বাধীন ইউনিটে বিভক্ত থাকে যাকে **Cell** বলা হয়। প্রতিটি সেল একটি বিচ্ছিন্ন দ্বীপের মতো চলে এবং অন্য সেলের ওপর কোনো নির্ভরতা থাকে না।
- **সেল রাউটার (Cell Router)**: একটি অত্যন্ত হালকা ও নির্ভরযোগ্য এজ রাউটার প্রতিটি ইনকামিং রিকোয়েস্টের টেন্যান্ট আইডি বা পার্টিশন হ্যাশ পরীক্ষা করে সেই নির্দিষ্ট সেলে ট্রাফিক পাঠিয়ে দেয়।
- **ব্লাস্ট রেডিয়াস নিয়ন্ত্রণ**: কোনো অপ্রত্যাশিত ডেটাবেস ক্র্যাশ বা সফটওয়্যার বাগ Cell 1-কে পুরোপুরি ধ্বংস করে দিলেও, **Cell 2, Cell 3, Cell 4 ... N শতভাগ সচল থাকে**। ফলে বিশ্বব্যাপী বিভ্রাটের বদলে মাত্র ৫% গ্রাহক প্রভাবিত হয়।
- **ট্রেড-অফ**: একাধিক সেলের সম্মিলিত ডেটা বা গ্লোবাল অ্যানালিটিক্স কুয়েরি করতে বিশেষ অ্যাসিঙ্ক্রোনাস ডেটা পাইপলাইনের প্রয়োজন হয়।

### প্রশ্ন ৭: স্পেস-বেসড আর্কিটেকচারে ডেটাবেসের বটলনেক সম্পূর্ণ দূর করার পাশাপাশি ডেটা লস কীভাবে শূন্যের কোঠায় নামিয়ে আনা যায়?
**উত্তর**:
স্পেস-বেসড আর্কিটেকচার সমস্ত লেনদেন র‍্যামের (RAM) মধ্যে রেপ্লিকেটেড **Processing Units**-এ সম্পন্ন করে ডেটাবেসের সেন্ট্রাল বটলনেক দূর করে:
১. **ইন-মেমোরি ডেটা গ্রিড**: সমস্ত রিড ও রাইট অপারেশনের প্রসেসিং মেমোরিতে সম্পন্ন হওয়ায় সাব-মিলিসেকেন্ড লেটেন্সি এবং অতি-উচ্চ থ্রুপুট অর্জিত হয়।
২. **সিঙ্ক্রোনাস ইন-মেমোরি রেপ্লিকেশন**: ক্লায়েন্টকে সফল রেসপন্স পাঠানোর আগেই মাস্টার নোডটি অন্তত একটি ভিন্ন অ্যাভেইলেবিলিটি জোনের স্ট্যান্ডবাই রেপ্লিকা নোডে ডেটা সিঙ্ক করে।
৩. **Write-Ahead Log (WAL) ও ডিস্ক রাইটিং**: প্রতিটি পরিবর্তন দ্রুতগতির NVMe স্টোরেজে অ্যাপেন্ড-অনলি ট্রানজ্যাকশন লগে লেখা হয়, এবং একটি অ্যাসিঙ্ক্রোনাস ব্যাচ রাইটার ইঞ্জিন ব্যাকগ্রাউন্ডে পারসিস্টেন্ট ডেটাবেসে ডেটা জমা করে।
৪. **.NET ইমপ্লিমেন্টেশন**: `Microsoft.Orleans` ভার্চুয়াল অ্যাক্টর ফ্রেমওয়ার্কের মাধ্যমে মেমোরি স্টেট ম্যানেজমেন্ট এবং অ্যাসিঙ্ক স্টোরেজ পারসিস্টেন্স (`IPersistentState<T>`) নিশ্চিত করা হয়।

### প্রশ্ন ৮: সার্ভারলেস (FaaS) আর্কিটেকচারের ট্রেড-অফ কী, এবং .NET-এ কোল্ড স্টার্ট ও ডেটাবেস কানেকশন এগজশন কীভাবে সমাধান করবেন?
**উত্তর**:
- **সুবিধাসমূহ**: শূন্য থেকে ১০,০০০+ কন্টেইনারে তাৎক্ষণিক স্কেলিং, ট্রাফিক শূন্য থাকলে শূন্য অবকাঠামো বিল (\$0), এবং সার্ভার ওএসের কোনো রক্ষণাবেক্ষণ না থাকা।
- **ট্রেড-অফ**: স্টেটলেস সীমাবদ্ধতা, প্রথম রিকোয়েস্টে কোল্ড স্টার্ট লেটেন্সি (১০০ms–২s), এবং ফাংশন এক্সিকিউশন টাইমআউট।
- **.NET 9-এ কোল্ড স্টার্ট দূর করার উপায়**:
  ১. **.NET Isolated Worker Model** (`Microsoft.Azure.Functions.Worker`) ব্যবহার করা।
  ২. **Native AOT (Ahead-of-Time compilation)** চালু করা, যা রানটাইম JIT ওভারহেড দূর করে স্টার্টআপ লেটেন্সি <৫০ মিলিসেকেন্ডে নামিয়ে আনে।
  ৩. অপ্রয়োজনীয় লাইব্রেরি ট্রিম করে প্যাকেজের আকার ক্ষুদ্র রাখা।
- **ডেটাবেস কানেকশন পুল সংকট সমাধান**: হঠাৎ ১,০০০ ফাংশন ইন্সটেন্স চালু হলে রিলেশনাল ডেটাবেসের কানেকশন লিমিট শেষ হয়ে যায়।
  - *সমাধান*: ক্লাউড কানেকশন প্রক্সি (যেমন **Azure SQL Database Hyperscale Serverless**, **AWS RDS Proxy**, বা **PgBouncer**) ব্যবহার করা, অথবা HTTP-ভিত্তিক সার্ভারলেস স্টোর যেমন Azure Cosmos DB বা DynamoDB ব্যবহার করা।

### প্রশ্ন ৯: ডিস্ট্রিবিউটেড সিস্টেমে "Dual-Write Problem" কেন ঘটে এবং ট্রানজ্যাকশনাল আউটবক্স প্যাটার্ন কীভাবে নির্ভরযোগ্য সমাধান দেয়?
**উত্তর**:
- **ডুয়াল-রাইট সমস্যা**: যখন কোনো ব্যবসায়িক লজিকে একই সাথে ডেটাবেসে স্টেট সেভ করতে হয় এবং মেসেজ ব্রোকারে ইভেন্ট পাঠাতে হয় (যেমন: অর্ডার সেভ করা এবং `OrderCreatedEvent` পাবলিশ করা), তখন এদের দুটি পৃথক নেটওয়ার্ক কলে সম্পন্ন করলে 2PC প্রোটোকল ছাড়া অ্যাটমিক গ্যারান্টি পাওয়া যায় না। ডেটাবেস রাইট সফল হওয়ার পর ব্রোকার ডাউন হলে বা অ্যাপ ক্র্যাশ করলে সিস্টেমে অসঙ্গতি তৈরি হয়।
- **সমাধান (Transactional Outbox)**:
  ১. একটিমাত্র লোকাল ACID ট্রানজ্যাকশনের মধ্যে বিজনেস এন্টিটি সেভ করার পাশাপাশি একই ডেটাবেসের `OutboxMessages` টেবিলে সিরিয়ালাইজড মেসেজ ইনসার্ট করা হয়।
  ২. দুটো অপারেশন একই রিলেশনাল ট্রানজ্যাকশনে থাকায় অ্যাটমিসিটি ১০০% নিশ্চিত হয়।
  ৩. একটি নির্ভরযোগ্য ব্যাকগ্রাউন্ড প্রসেস (যেমন MassTransit Outbox Worker বা Postgres WAL রিড করা Debezium CDC) আউটবক্স টেবিল থেকে মেসেজ পড়ে **At-least-once delivery** গ্যারান্টিসহ ব্রোকারে পাঠায় এবং মেসেজটিকে প্রসেসড হিসেবে চিহ্নিত করে।

### প্রশ্ন ১০: ইভেন্ট-ড্রিভেন সিস্টেমে কখন ব্রোকার টপোলজি (Choreography) এবং কখন মিডিয়েটর টপোলজি (Orchestration) বেছে নেবেন?
**উত্তর**:
- **ব্রোকার টপোলজি (Choreography)**:
  - *কার্যপদ্ধতি*: সার্ভিসগুলো ব্রোকারে ইভেন্ট প্রকাশ করে; ডাউনস্ট্রিম কনজিউমাররা কোনো কেন্দ্রীয় সমন্বয়কারী ছাড়াই স্বাধীনভাবে ইভেন্ট গ্রহণ ও প্রতিক্রিয়া জানায়।
  - *সুবিধা*: সর্বোচ্চ শিথিল সংযোগ (Loose Coupling), উচ্চ থ্রুপুট, কোনো সেন্ট্রাল বটলনেক নেই।
  - *অসুবিধা*: সামগ্রিক ওয়ার্কফ্লোর অবস্থা কোথাও কেন্দ্রীয়ভাবে সংরক্ষিত থাকে না; জটিল ব্যর্থতা পর্যবেক্ষণ ও ডিবাগ করা কঠিন।
  - *ব্যবহার*: নোটিফিকেশন সিস্টেম, রিয়েল-টাইম মেট্রিক্স, টেলিমেট্রি এবং স্বাধীন ডেটা ব্রডকাস্ট।
- **মিডিয়েটর টপোলজি (Orchestration)**:
  - *কার্যপদ্ধতি*: একটি কেন্দ্রীয় অর্কেস্ট্রেটর (যেমন MassTransit State Machine, Temporal বা Azure Durable Functions) প্রতিটি সার্ভিসকে ধাপে ধাপে নির্দেশ দেয় এবং তাদের উত্তরের ওপর ভিত্তি করে পরবর্তী সিদ্ধান্ত গ্রহণ করে।
  - *সুবিধা*: সম্পূর্ণ লেনদেনের অবস্থা পর্যবেক্ষণযোগ্য, ব্যর্থ হলে কম্পেনসেটিং রোলব্যাক (Saga) পরিচালনা করা সহজ।
  - *অসুবিধা*: অর্কেস্ট্রেটর সিস্টেমের জন্য একটি একক ব্যর্থতার বিন্দু (SPOF) এবং বটলনেকে পরিণত হতে পারে।
  - *ব্যবহার*: ই-কমার্স পেমেন্ট-অর্ডার-শিপিং চেকআউট সাগা এবং মাল্টি-স্টেপ ব্যাংকিং লেনদেন।

### প্রশ্ন ১১: লেয়ারড আর্কিটেকচারে "Architecture Sinkhole" সমস্যা কী এবং কীভাবে এটি দূর করবেন?
**উত্তর**:
- **সমস্যা**: ট্র্যাডিশনাল লেয়ারড আর্কিটেকচারে (Presentation $\rightarrow$ Business $\rightarrow$ Data Access) অধিকাংশ অপারেশনই (প্রায় ৬০–৮০%) সাধারণ রিড-অনলি কুয়েরি। যখন কোনো রিকোয়েস্ট প্রতিটি লেয়ারের মধ্য দিয়ে যায় কিন্তু কোনো বিজনেস ভ্যালিডেশন বা রূপান্তর ঘটে না, তখন তাকে "সিঙ্কহোল" বলে—যা শুধু অর্থহীন বয়লারপ্লেট কোড এবং মেমোরি অ্যালোকেশন বৃদ্ধি করে।
- **সমাধান**:
  ১. **রিল্যাক্সড লেয়ারিং (Relaxed Layering)**: কেবল রিড অপারেশনের জন্য প্রেজেন্টেশন লেয়ারকে সরাসরি ডেটা অ্যাক্সেস লেয়ার কল করার অনুমতি দেওয়া, যাতে সার্ভিস লেয়ার শুধু বিজনেস রাইট অপারেশনের দায়িত্ব সামলায়।
  ২. **ভার্টিক্যাল স্লাইস আর্কিটেকচার (Vertical Slice)**: অনুভূমিক লেয়ারের বদলে ফিচারভিত্তিক স্লাইসে কোড সাজানো (যেমন `Features/Orders/GetOrderById`)। MediatR বা FastEndpoints ব্যবহার করে কোনো অতিরিক্ত লেয়ার ছাড়াই সরাসরি Dapper বা EF Core দিয়ে ডেটা রিটার্ন করা।

### প্রশ্ন ১২: Entity Framework Core-এর ওপর জেনেরিক রিপোজিটরি এবং ইউনিট অফ ওয়ার্ক তৈরি করা কি একটি অ্যান্টি-প্যাটার্ন?
**উত্তর**:
- **প্যারাডক্স**: EF Core-এর `DbContext` নিজেই **Unit of Work** প্যাটার্নের একটি নিখুঁত বাস্তবায়ন (এটি এন্টিটির স্টেট ট্র্যাক করে এবং `SaveChangesAsync()`-এর মাধ্যমে একবারে ট্রানজ্যাকশন কমিট করে), এবং `DbSet<T>` নিজেই একটি ইন-মেমোরি **Repository** যা Linq কুয়েরি সাপোর্ট করে।
- **অ্যান্টি-প্যাটার্ন**: `GetAll()`, `GetById()`, `Add()` সহ এমন একটি জেনেরিক `IRepository<T>` তৈরি করা যা বাইরে `IQueryable<T>` এক্সপোজ করে—তা একটি ত্রুটিপূর্ণ অ্যাবস্ট্রাকশন (Leaky Abstraction)। অ্যাপ্লিকেশন লেয়ারে লেখা জটিল LINQ লজিক অবকাঠামোর বাইরে ছড়িয়ে পড়ে।
- **স্টাফ ইঞ্জিনিয়ারের বাস্তবায়ন**:
  ১. জেনেরিক রিপোজিটরি কখনোই ব্যবহার না করে **ডোমেইন-নির্দিষ্ট রিপোজিটরি** (যেমন `IOrderRepository`) তৈরি করা যাতে কেবল সুনির্দিষ্ট বিজনেস মেথড (যেমন `Task<Order?> GetOrderWithItemsAsync(Guid id)`) থাকে।
  ২. ক্লিন আর্কিটেকচারে ডোমেইন-নির্দিষ্ট রিপোজিটরি ইন্টারফেসগুলো কোর অ্যাপ্লিকেশনের **Outbound Port** হিসেবে কাজ করে, আর EF Core-এর `DbContext` ইনফ্রাস্ট্রাকচার লেয়ারের **Outbound Adapter** হিসেবে কাজ করে—যার ফলে ডোমেইন লেয়ার EF Core লাইব্রেরি থেকে ১০০% মুক্ত থাকে।

### প্রশ্ন ১৩: CQRS এবং ইভেন্ট সোর্সিং কীভাবে কঠোর নিয়ন্ত্রক কমপ্লায়েন্স (SOC2/GDPR) নিশ্চিত করে এবং স্কিমা বিবর্তন কীভাবে সামাল দেয়?
**উত্তর**:
- **কমপ্লায়েন্স ও অডিট ট্রেইল**:
  - ঐতিহ্যবাহী CRUD মডেলে একটি `UPDATE` কুয়েরি পূর্বের সমস্ত স্টেট মুছে ফেলে, যার ফলে কে কখন কী তথ্য পরিবর্তন করেছে তা জানা অসম্ভব।
  - **ইভেন্ট সোর্সিং**-এ স্টেট কোনো সাধারণ টেবিল নয়, বরং অপরিবর্তনীয় ডোমেইন ইভেন্টের ধারাবাহিক লগ (`OrderPlaced`, `AddressChanged`, `PriceOverridden`)। প্রতিটি ইভেন্টে ব্যবহারকারীর আইডি, টাইমস্ট্যাম্প ও আইপি সংরক্ষিত থাকে—যা SOC2 ও ব্যাংকিং অডিটের জন্য গাণিতিকভাবে নিখুঁত হিস্ট্রি প্রদান করে।
  - **CQRS** প্রতিটি কমান্ডের জন্য আলাদা পারমিশন চেক করার সুযোগ দেয় (`ChangeAddressCommand` বনাম `ApproveCreditLimitCommand`), যা কোনো ইউজারকে অপ্রয়োজনীয় আপডেট সুবিধা পাওয়া থেকে বিরত রাখে।
- **স্কিমা বিবর্তন (Schema Evolution)**:
  ১. *Tolerant Reader*: ডিসিরিয়ালাইজার নতুন যুক্ত হওয়া ফিল্ড উপেক্ষা করে এবং পুরনো লগের ক্ষেত্রে ডিফল্ট মান বসায়।
  ২. *Upcasters*: একটি ইন-মেমোরি পাইপলাইন পুরনো ইভেন্ট ভার্সনকে (যেমন `UserRegisteredV1`) মেমোরিতে স্বয়ংক্রিয়ভাবে নতুন মডেলে (`UserRegisteredV2`) রূপান্তর করে ডোমেইন হ্যান্ডলারকে প্রদান করে।
  ৩. *GDPR ক্রিপ্টো-শ্রেডিং (Crypto-Shredding)*: গ্রাহকের ব্যক্তিগত তথ্য (PII) একটি নির্দিষ্ট ক্রিপ্টোগ্রাফিক কি দিয়ে এনক্রিপ্ট করে ইভেন্ট লগে রাখা হয়। গ্রাহক ডেটা মুছে ফেলতে চাইলে কেবল সেই এনক্রিপশন কি মুছে ফেলা হয়—যার ফলে সম্পূর্ণ ইভেন্ট লগ অপরিবর্তিত রেখেই গ্রাহকের তথ্য চিরতরে অপাঠ্য হয়ে যায়।

### প্রশ্ন ১৪: ওয়েব ও মোবাইল ফ্রন্টএন্ডে MVC/MVVM থেকে Unidirectional Data Flow (MVI/Flux) কেন জনপ্রিয় হলো, এবং BFF প্যাটার্ন কী সমাধান করে?
**উত্তর**:
- **ইউআই স্টেটের বিবর্তন**:
  - **MVC / MVP**: কন্ট্রোলার সরাসরি ভিউ পরিবর্তন করত, যার ফলে কোডবেস টাইটলি কাপল্ড এবং টেস্ট করা অসম্ভব হয়ে উঠত।
  - **MVVM**: টু-ওয়ে ডেটা বাইন্ডিং যুক্ত করে। কিন্তু জটিল স্ক্রিনে দ্বিমুখী বাইন্ডিংয়ের ফলে ভিউ ও ভিউমডেলের মধ্যে অনাকাঙ্ক্ষিত "পিং-পং" স্টেট লুপ তৈরি হতো, যার ফলে বাগ ট্র্যাক করা দুঃসাধ্য ছিল।
  - **MVI / Redux (Unidirectional Data Flow - UDF)**: একমুখী স্টেট চক্র প্রয়োগ করে: `User Intent -> Action/Command -> Reducer/State Machine -> Immutable State -> Rendered View`। স্টেট অপরিবর্তনীয় হওয়ায় টাইম-ট্রাভেল ডিবাগিং এবং ত্রুটি নির্ভুলভাবে পুনরুৎপাদন করা সম্ভব হয়।
- **Backend-for-Frontend (BFF) প্যাটার্ন**:
  - একটি ডিস্ট্রিবিউটেড ব্যাকএন্ডে মোবাইল অ্যাপ, ডেস্কটপ ব্রাউজার এবং থার্ড-পার্টি পার্টনারদের ব্যান্ডউইথ ও ডেটা চাহিদায় বিশাল পার্থক্য থাকে।
  - একটিমাত্র সাধারণ API Gateway ব্যবহার করলে দুর্বল মোবাইল নেটওয়ার্কে ক্লায়েন্ট অপ্রয়োজনীয় মেগাবাইট ডেটা (Over-fetching) ডাউনলোড করতে বাধ্য হয়।
  - **BFF প্যাটার্ন** প্রতিটি ফ্রন্টএন্ডের জন্য আলাদা লাইটওয়েট গেটওয়ে তৈরি করে (যেমন `Web-BFF`, `Mobile-BFF`)। BFF ব্যাকএন্ড মাইক্রোসার্ভিস থেকে প্রয়োজনীয় ডেটা সংগ্রহ করে, অপ্রয়োজনীয় ফিল্ড ছেঁটে ফেলে এবং ক্লায়েন্টের ডিসপ্লে উপযোগী ফরম্যাটে একবারে রেসপন্স পাঠায়।

### প্রশ্ন ১৫: ডিস্ট্রিবিউটেড সিস্টেমে ক্যাসকেডিং ফেইলিউর রোধে সার্কিট ব্রেকার, এক্সপোনেনশিয়াল ব্যাকঅফ উইথ জিটার এবং বাল্কহেড কীভাবে কাজ করে?
**উত্তর**:
ডিস্ট্রিবিউটেড সিস্টেমে কোনো একটি ডাউনস্ট্রিম মাইক্রোসার্ভিসের ধীরগতি সম্পূর্ণ এন্টারপ্রাইজ সিস্টেমের থ্রেড ও মেমোরি দখল করে পুরো আর্কিটেকচার ক্র্যাশ করাতে পারে:
১. **Exponential Backoff with Jitter**:
   - কোনো সার্ভিস ফেইল করলে সাথে সাথে বারবার কল করলে রিকভার করার আগেই সার্ভিসটি রিকোয়েস্টের বন্যায় ধ্বংস হয়ে যায় (Thundering Herd Problem)।
   - র্যান্ডম জিটার যুক্ত ব্যাকঅফ ($T_{wait} = 2^{attempt} \times \text{Base} + \text{Random}()$) ক্লায়েন্টদের রিট্রাই রিকোয়েস্টগুলোকে ছড়িয়ে দেয়, ফলে ডাউনস্ট্রিম সার্ভিস নিরাপদে রিকভার করার সুযোগ পায়।
২. **Circuit Breaker (Polly / `Microsoft.Extensions.Http.Resilience`)**:
   - তিনটি স্টেটে চলে: **Closed** (স্বাভাবিক), **Open** (ক্রমাগত ফেইল করলে সরাসরি তাৎক্ষণিক ফল্ট ফেরত দেয় কোনো নেটওয়ার্ক কল ছাড়াই), এবং **Half-Open** (সার্ভিস ঠিক হয়েছে কিনা তা পরীক্ষা করতে সীমিত রিকোয়েস্ট পাঠায়)।
   - এটি ক্লায়েন্টের থ্রেড পুল ও কানেকশন রিসোর্সগুলো বাঁচিয়ে রাখে।
৩. **বাল্কহেড আইসোলেশন (Bulkhead Isolation)**:
   - জাহাজের নিচের প্রকোষ্ঠগুলোর মতো: একটি প্রকোষ্ঠে পানি ঢুকলেও পুরো জাহাজ ডোবে না।
   - কোনো নির্দিষ্ট বহিঃস্থ সার্ভিসের জন্য সর্বোচ্চ সমান্তরাল থ্রেড বা কানেকশন কোটা বরাদ্দ করে দেয় (যেমন: Payment API-এর জন্য সর্বোচ্চ ২০টি থ্রেড)। পেমেন্ট সার্ভিস সম্পূর্ণ হ্যাং করলেও কেবল সেই ২০টি থ্রেড ব্লক হবে; অ্যাপ্লিকেশনের বাকি শত শত থ্রেড প্রোডাক্ট ব্রাউজিং, সার্চ বা লগইন রিকোয়েস্ট স্বাভাবিকভাবে পরিচালনা করতে থাকবে।
