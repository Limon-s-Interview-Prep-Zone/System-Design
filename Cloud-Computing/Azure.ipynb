### Azure Storage Services
Azure storage services are a suit of cloud storage solutions provided by Microsoft Azure. These services offer durable and secure storage options for various types of data, ranging from simple text to media files and even big databases.
- The allow users to store, retrieve, and manage data in Azure.
- Provide flexibility and scalability to meet various storage needs.
- **`Type:`** Blobs, Files, Queues, Tables, Disk, Data Lake, Archive
- Except from disk, every other service requires setting up a storage account to begin using them.
- ***`Storage Account:`*** A storage is like a Virtual container for your data, providing ways to organize, access and secure your data in cloud.


#### Blobs(Binary large Objects) Storage:
Blobs, or Binary Large Objects, are designed for storing large amounts of unstructured data, such as documents, images, videos, and others.<br>
Key Features:
1. `Scalability`: Can store petabytes of data.
2. `High Availability`: Redundant copies stored across regions for fault tolerance.
3. `Access Tiers`: Offers Hot, Cool, and Archive tiers to optimize costs.


Use-Case:
- Storing files for widespread access
- Delivery of images or documents to web
- `Media Storage`: Store images and videos for streaming applications.
- `Backups & Disaster Recovery`: Store application backups, disater recovery, archives, and log files.
- `Big Data Analytics`: Store raw data before processing with Azure Data Lake.

Trade-offs:
- Not suitable for `structured data`, `higher latency` compared to other storage types
- `No File System`: Unlike traditional file storage, it doesn’t have a directory structure.
- `Example:` A news website stores `user-uploaded images and videos` in Blob Storage, using the` Hot tier for frequently accessed content` and `the Cool tier for archived files`.

##### Tiers
1. Hot Access Tier:
   - Frequently access data like images on a website.
2. Cool Access Tier:
   - Data access for infrequently, such as customer invoice
3. Cold Access Tier:
   - Rarely access data
4. Archive Tier: It is designed for achiving like long-term backups.

#### Azure Files storage service:
Azure files service provides organized file spaces in the cloud that can be easily accessed from various devices, including both cloud and on-premises systems. Once connected to Azure cloud, files can be accessed from most `operating systems, including Windows, Linux, and macOS.`<br>

Key Features:
- `Network File System`: Mount file shares across multiple `virtual machines`.
- `Supports Windows and Linux`: Accessible via SMB for Windows and NFS for Linux.
- `Fully Managed`: No need to manage hardware or OS updates.


Use Cases:
- `Distributed File Storage`: When a shared storage solution is required for multiple users or applications such as `Sharing files between virtual machines`
- The ability to access files from various operating systems, when cross-platform access is needed.
- You require a centralized cloud location for file collaboration. 
- You need to share application data between different instances or components of an application such as `Configuration files`.
- `Lift-and-Shift Applications`: Migrate legacy applications that need shared storage.

Trade-offs:
- Limited scalability compared to Blob Storage, higher cost for large-scale storage
- `Example`: A company migrates an on-premises file server to Azure Files, allowing remote employees to access shared company documents from anywhere.

#### Queue:
Queue storage is a digital message service in the cloud. This messaging system enables communication between various components, programs or applications, helping them work together efficiently by sharing information and instructions. These messages are stored, basically creating a queue of working instructions. Usually, queue services are implemented in workflows containing multiple other services.

Key Features:
-` Message Queues`: Store millions of messages up to 64 KB in size.
- `Asynchronous Processing`: Allows distributed applications to communicate efficiently.
- `Durable and Scalable`: Messages persist until they are processed.

Use Cases:
- `Event-Driven Processing`: Trigger actions when new data arrives.
- `Load Balancing`: Distribute tasks across multiple workers.
- `Microservices Communication`: Enables loosely coupled microservices

Trade-offs:
- `Limited FIFO Support`: Basic queuing, not advanced message ordering.
- `Smaller Message Size`: Not suitable for large payloads.
- `Example`: An e-commerce app uses Queue Storage to manage order processing, ensuring customer orders are handled asynchronously without overloading the system.

#### Tables:
Azure Table Storage is a NoSQL key-value store that allows for fast, structured storage of non-relational data.

Key Features:
- `Schema-less Design`: Store structured data without a fixed schema.
- `Scalability`: Supports massive datasets with fast lookups.
- `Low Cost`: Cost-effective compared to traditional databases.
- `Example`: An IoT system stores temperature readings from thousands of sensors in Azure Table Storage for real-time monitoring.

Use Cases:
- `Logging and Telemetry`: Store logs and performance metrics.
- `User Profiles`: Store non-relational data like user settings and preferences.
- `Session Management`: Save session state for web applications.

Trade-offs:
- No Complex Queries: Doesn’t support SQL-style joins.
- Limited Indexing: Only indexed on primary keys.


#### Disks
Azure Disks, also known as Azure Managed Disks, provide scalable and secure persistent storage for virtual machines.  They are similar to regular hard drives but in a virtual form. It allows you to attach disk volumes to your virtual machines, providing them with the necessary storage space for operating systems, applications, and data. Azure Disks play a crucial role in virtual machines functionality.
Key Features:
- `Managed Disks`: Fully managed by Azure, supporting `SSDs and HDDs`.
- `High Performance`: Supports up to 160,000 IOPS (Input/Output Operations Per Second).
- `Data Protection`: Built-in backup and replication options.

Use Cases:
- `Virtual Machine Storage`: Attach disks to VMs for high-speed performance.
- `Databases`: Store structured data with high availability.
- `Enterprise Applications`: SAP, Oracle, and other enterprise workloads.

Trade-offs:
- `Expensive for Large Storage Needs`: Costs more than Blob and File storage.
- `Limited Access`: Only accessible via attached VMs.
- Example: A financial services company uses Azure Premium SSDs for its SQL database to ensure high-speed transactions and minimal downtime.

#### Azure Data Lake Storage
Azure Data Lake Storage is a scalable data storage and analytics service designed for big data workloads. It supports massive amounts of data and integrates with various analytics frameworks.


Trade-offs:
- Pros: Highly scalable, supports hierarchical namespace, integrates with big data tools.
- Cons: Higher complexity and cost compared to Blob Storage, requires expertise in big data technologies.

Use Case: A data science team storing and analyzing petabytes of data for machine learning models.

#### Storage Configurations and file managements

##### Storage Redundency:
1. ***`LRS(Local Redundant Storage)`***: Stores `three copies` of data within a single data center.
   - Lowest cost but no disaster recovery.
2. ***`ZRS(Zone Redundant Storage)`***: Replicates data across `three availability zones` in the same region.
   - Copy 1: Stored in Availability Zone 1 (AZ1), Copy 2: Stored in Availability Zone 2 (AZ2). Copy 3: Stored in Availability Zone 3 (AZ3).
   - High availability within a region but no regional failover.
3. ***`GRS(Geo Redundant Storage)`***: Replicates data to a secondary region (read-only).
   - each primary and secondary data center contains `LRS(3 copies of data)`
   - Higher cost, slower failover.
   - The secondary region is read-only unless you perform a failover.
   - Provides disaster recovery but with higher latency for secondary access.
4. ***`GZRS(Geo-Zone Redundant Storage)`***: Similar to GRS but allows read access to the secondary region.
   - Data copied in `3 availability zones` within 3 different data centers in the primary zones and also replicated in secondary region(`LRS`).
   - Higher cost but ensures minimal downtime during regional failures.

#### Azure Storage Lifecycle Management Policies
Azure Storage Lifecycle Management allows you to automatically move, delete, or expire data in Blob Storage based on predefined rules. This helps optimize storage costs by transitioning data between access tiers (Hot, Cool, Archive) and deleting unnecessary files.

##### Why Use Lifecycle Management?
- `Cost Optimization`: Move infrequently accessed data to cheaper storage tiers.
- `Automated Data Retention`: Ensure compliance by retaining data for a fixed period.
- `Reduce Manual Effort`: No need to manually monitor and move blobs.
- `Policy-Driven Storage Management`: Set rules based on last access or modification time.

Common Use Cases & Trade-offs<br>
| Use Case                   | Lifecycle Policy                                     | Trade-offs                                                     |
|----------------------------|-----------------------------------------------------|---------------------------------------------------------|
| Log Data Archival          | Move blobs to Cool storage after 30 days, then delete after 365 days. | + Cost savings on storage. <br> 🔹 May have higher access latency when needed. |                                                   |
| Data Compliance (Long-term storage) | Move blobs to Archive after 90 days, delete after 1 year.     | + Ensures data is stored as per compliance rules. <br> - Higher retrieval time from Archive. |                                                   |
| Temporary File Cleanup     | Delete temporary files older than 7 days.           | + Reduces storage costs. <br> - Risk of accidental data loss. |                                                   |
| Tiering for cold data      | Move from Hot → Cool → Archive based on last modified date. | + Saves costs. <br> - Retrieval from Archive can be slow (hours delay and rehydration cost). |                                                   |
| Retention for compliance   | Automatically delete blobs after a set retention period. | + Reduces storage costs and ensures data retention policies are met. |                                                   |

## 2#Compute and Networking Service

### Azure Compute Service
Azure compute services provide the necessary computing power to run apps, process data, and perform various tasks.

#### Azure Serverless Computing:
Azure serverless computing refers to a broder variety of services that allow to build and run application without managing underlying infrastructure.
1. Azure Functions (event-driven compute)
2. Azure Logic Apps (workflow automation)
3. Azure Event Grid (event-based messaging)
4. Azure Container Apps (serverless container management)

##### 1#Azure Function:
Azure Functions is a serverless compute service that allows you to run event-driven code without managing infrastructure. It automatically scales based on demand and charges only for the execution time.<br>

Use-case:
- Automating workflows (e.g., process files, trigger actions on database updates)
- Building APIs (e.g., backend services for mobile/web apps)
- IoT event processing (e.g., processing sensor data from IoT devices)
- Scheduled tasks (e.g., running nightly reports)

**Tradeoffs**:
1. Pros:
   - No server management: Azure handles infrastructure, scaling, and maintenance.
   - Cost-effective: Pay only for the time your code runs.
   - Event-driven: Ideal for scenarios triggered by events (e.g., HTTP requests, database changes, or messages).
   - Rapid development: Quick to deploy and integrate with other Azure services.
2. Cons:
   - `Cold starts`: Functions may experience latency when starting after inactivity. When a serverless function hasn't been used for a while, it may take some time to start up (cold start), leading to latency.
   - `Limited execution time`: Functions have a maximum execution time (default is 5 minutes, configurable up to 10 minutes).
   - `Stateless`: Not ideal for long-running or stateful processes.
   - Harder to trace issues across multiple functions and services.
---
##### 2#Azure Logic Apps:
Azure Logic Apps is a cloud-based workflow automation service that connects different services using a visual designer. It enables low-code/no-code automation.

Use Cases:
- Business process automation (e.g., approval workflows, CRM updates)
- Data integration (e.g., syncing data between databases)
- Automating notifications (e.g., sending alerts on system failures)
- Connecting SaaS applications (e.g., Salesforce, Office 365, SAP)
- Example: When an email arrives in a specific inbox, a Logic App workflow extracts details, logs them into a database, and notifies the relevant Slack channel.


Trade-offs:
- ✅ No-code workflow design (easy for non-developers)
- ✅ Pre-built connectors for hundreds of services
- ✅ Built-in monitoring and error handling
- ❌ Higher cost for frequent executions
- ❌ Limited customization compared to Azure Functions
- ❌ Can have execution delays for complex workflows

---
##### 3#Azure event Grid:
Azure Event Grid is a messaging service for handling event-based communication between applications and services. It supports pub/sub (publish/subscribe) messaging.

Use Cases:
- Event-driven applications (e.g., triggering actions on file uploads)
- Microservices communication (e.g., event-driven architecture)
- Real-time notifications (e.g., alerting users when stock prices change)
- Serverless automation (e.g., triggering workflows on database updates)
- A healthcare system uses Event Grid to notify different services when new patient records are added. It triggers workflows to update insurance systems, send appointment reminders, and sync data across multiple departments.


Trade-offs:
- ✅ Highly scalable event processing
- ✅ Supports multiple event sources (Blob Storage, Azure Functions, etc.)
- ✅ Low-latency message delivery
- ❌ Requires event handlers (Azure Functions, Logic Apps, or custom subscribers)
- ❌ Message duplication possible (needs idempotency handling)
- ❌ Can be complex to debug

---
#### 4#Azure Container Apps:
Azure Container Apps is a fully managed, serverless container runtime for running microservices and containerized applications without managing Kubernetes infrastructure.

Use Cases:
- Running microservices without managing Kubernetes
- Deploying APIs and web apps in a serverless model
- Event-driven processing for containerized applications
- Scaling stateless workloads dynamically


Trade-offs:
- ✅ Auto-scaling based on events (HTTP, CPU, memory, etc.)
- ✅ Runs standard containers (supports any language/runtime)
- ✅ Integrated with Dapr for service-to-service communication
- ❌ More expensive than Azure Functions for simple tasks
- ❌ Cold starts on infrequent executions
- ❌ Less mature than Kubernetes (AKS) for complex apps

### Azure Networking
Network services are a fundamental component of Azure cloud, providing the ability to connect and secure cloud resources. Understanding networking components is vital for designing and managing cloud-based solutions.

#### Cloud Resource Connectivity
Cloud Resource Connectivity refers to how different Azure resources communicate securely and efficiently across different networks, services, and environments.
1. Azure VirtualNet(Vnet)
2. Azure VPN Gateway- Hybrid cloud connectivity (cheaper option)
3. Azure Express Route- High-speed private connection for enterprises
4. Azure Application Gateway - Secure web traffic management
5. VNet peering- Private connection between VNets
6. Azure CDN – Faster content delivery
7. Azure DNS – Domain name management
8. Azure Private Link


##### 1#Azure VirtualNet(Vnet)
Azure Virtual Network (VNet) is like your own private data center in Azure, where your virtual machines (VMs), databases, and other resources can communicate securely.<br>
Use-Case:
- ✅ Secure communication between Azure resources (e.g., VMs, databases, Kubernetes clusters)
- ✅ Hybrid networking (connect on-premises data center to Azure using VPN)
- ✅ Network segmentation (create subnets for better security and organization)
- Example: A bank deploys web, app, and database tiers in Azure. Each tier is placed in a separate subnet within a VNet for security and isolation. The web tier is accessible over the internet, while the app and database tiers are private.


⚖️ Trade-offs
- ✔️ Secure and isolated from the public internet
- ✔️ Customizable with subnets, security rules, and firewalls
- ❌ Not publicly accessible by default (must set up VPN, ExpressRoute, or NAT Gateway)

##### 2#Azure VPN Gateway:
Azure VPN Gateway connects on-premises networks to Azure over secure VPN tunnels (site-to-site or point-to-site). It’s a cost-effective option for hybrid cloud connectivity.<br>
Use Cases:
- Secure connectivity between on-premises data centers and Azure.
- Remote user access to Azure resources.
- Example: A retail company connects its on-premises data center to Azure using a site-to-site VPN. This allows seamless integration of on-premises inventory systems with Azure-based e-commerce platforms. Employees access internal apps from home using a point-to-site VPN.

Trade-offs:
- Pros: Secure, easy to set up, and cost-effective.
- Cons: Limited bandwidth compared to ExpressRoute.

##### 3#Azure Express Route
ExpressRoute provides a dedicated, private connection between on-premises data centers and Azure, bypassing the internet.<br>
Use Cases
- ✅ Low-latency, high-speed connections for mission-critical workloads
- ✅ Hybrid cloud deployments (data center-to-Azure connectivity)
- ✅ Compliance-driven industries (e.g., finance, healthcare) requiring private networking
- Example: A financial institution uses ExpressRoute to connect its on-premises trading systems to Azure for real-time data analysis and reporting.

⚖️ Trade-offs
- ✔️ Faster and more reliable than VPN (up to 100 Gbps speeds)
- ✔️ More secure (avoids internet exposure)
- ❌ Expensive (monthly costs + setup fees)
- ❌ Requires physical setup with a network provider

##### 4#Azure Application Gateway
Azure Application Gateway is a Layer-7 (HTTP/HTTPS) load balancer with web application firewall (WAF) capabilities. It routes traffic based on URL paths or hostnames and provides SSL termination.<br>
Use-Case:
- ✔️ Web applications needing intelligent routing (e.g., /api goes to API servers).
- ✔️ E-commerce platforms & SaaS companies.
- ✔️ Securing apps against cyber threats (WAF protection).
- Example: A SaaS company has multiple web applications (app1.com, app2.com), and Application Gateway routes traffic based on the URL.


Trade-offs:
- Pros: Advanced routing and WAF capabilities.
- Cons: Higher cost compared to Layer-4 load balancers.

##### 5#Azure VNet peering
VNet Peering enables private connectivity between two Azure Virtual Networks (VNets). It allows resources in different VNets to communicate securely without traversing the public internet.<br>
Use-Case:
- ✔️Connecting VNets in the same or different regions.
- ✔️Enabling communication between applications hosted in separate VNets.
- ✔️ Microservices architectures where different services are in different VNets.
- ✔️ Connecting apps & databases securely across VNets.
- ✔️ Isolating production from development but allowing selective communication.
- Example:
  - VNet1: Production (web servers & databases).
  - VNet2: Testing (new deployments & QA).
  - VNet Peering allows seamless data flow between them.
Trade-Offs:
- Pros: Low latency, secure, and easy to configure.
- Cons: Peered VNets cannot have overlapping IP address ranges.


##### 6#Azure CDN
Azure CDN is a global network for delivering high-bandwidth content (e.g., images, videos) with low latency. It caches content at edge locations closer to users.
- Accelerating static content delivery.
- Reducing load on origin servers.
- An e-learning platform uses Azure CDN to deliver video lectures to students worldwide, ensuring smooth playback and reducing server load.
  
Trade-offs:
- Pros: Improved performance, reduced latency.
- Cons: Additional cost for high-bandwidth usage.
##### 7#Azure DNS
Azure DNS is a hosting service for DNS domains. It provides fast and reliable domain name resolution with high availability.
- Managing domain names for Azure-hosted services.
- Faster & more reliable than traditional DNS providers.
- Supports internal applications that don’t use public internet.

Trade-offs:
- Pros: High availability, integrated with Azure services.
- Cons: Limited to DNS management (not a full domain registrar).

##### 8#Azure Private Link:
Azure Private Link enables private access to Azure PaaS services (e.g., Azure Storage, Azure SQL Database) and customer-owned services over a private endpoint within your VNet.
- Securely accessing Azure PaaS services (e.g., Storage, SQL Database) from a VNet.
- Exposing your own services privately to other VNets or on-premises networks.
- Meeting compliance requirements by avoiding public internet exposure.
- A finance company needs a secure database but does NOT want to expose Azure SQL over the public internet → Private Link allows direct access via private IP.

Trade-offs:
- Pros: Enhanced security, private connectivity, and compliance.
- Cons: Limited to specific Azure services, additional cost for private endpoints.

### Azure Virtual Machine
An Azure Virtual Machine is an Infrastructure-as-a-Service (IaaS) and a cloud-based computing instance that runs an operating system (Windows or Linux) and provides compute, storage, and networking resources. Azure VMs allow users to deploy and manage applications without maintaining physical hardware.
- On-demand provisioning – Quickly create and delete VMs.
- Choice of OS – Supports Windows Server, Linux (Ubuntu, CentOS, Red Hat, etc.).
- Scalability – Scale up (increase size) or scale out (add instances).
- Custom Images – Use Azure Marketplace images or create custom VM images.
- Integration with Azure Services – Connect with databases, storage, security, and networking solutions.
- Backup & Disaster Recovery – Supports Azure Backup & Site Recovery.


#### Virtual Machine Components
1. Compute Components
    - vCPU & RAM – Defines the processing power and memory of the VM.
    - VM Size – Determines the number of vCPUs, memory, storage, and network capacity.
    - Operating System (OS) – Windows or Linux distributions.
2. VM Size: Azure offers various VM sizes optimized for different workloads:
   - General-purpose: Balanced CPU and memory (e.g., B-series, D-series).
   - Compute-optimized: High CPU performance (e.g., F-series).
   - Memory-optimized: High memory-to-CPU ratio (e.g., E-series, M-series).
   - Storage-optimized: High disk throughput (e.g., L-series).
   - GPU-optimized: For AI, machine learning, and graphics (e.g., NC-series, NV-series).
3. Storage Components
    - OS Disk – Persistent storage for the operating system.
    - Data Disks – Additional disks attached to the VM for application data.
    - Temporary Disk – A volatile disk that is cleared after a reboot.
    - Managed Disks – Azure-managed Standard HDD, Standard SSD, Premium SSD, Ultra Disk for better performance.
4. Networking Components
    - Virtual Network (VNet) – Connects VMs securely within Azure.
    - Network Interface (NIC) – Provides network connectivity to the VM.
    - Public & Private IP Addresses – Defines how the VM is accessed (internet vs. internal).
    - Network Security Group (NSG) – Controls inbound/outbound traffic.
    - Azure Load Balancer – Distributes traffic among multiple VM instances.
5. Security & Management Components
    - Azure Bastion – Provides secure RDP/SSH access without exposing a public IP.
    - Azure Defender – Protects against cyber threats.
    - Azure Monitor & Log Analytics – Tracks performance, usage, and alerts.
    - Azure Backup & Site Recovery – Ensures disaster recovery and business continuity.
6. Availability Options:
   - Availability Sets: Distributes VMs across fault domains and update domains for high availability.
   - Availability Zones: Distributes VMs across data centers within a region for higher resilience.
   - Scale Sets: Automatically scales the number of VMs based on demand.
#### Virtual Machine Usecase
- Use VMs for full control over the server environment.
- Host websites and web applications on Azure VMs.
- Run legacy or custom enterprise applications that require full control over the OS and infrastructure.
- Replicate on-premises VMs to Azure for backup and disaster recovery.
#### Scaling Virtual Machine
1. Vertical Scaling (Scaling Up/Down): Increasing or decreasing the size (CPU, memory, storage) of a single VM.
   -  Ideal for workloads with predictable resource requirements.
   -  Limitations:
      -  Requires downtime for resizing.
      -  Limited by the maximum VM size.
2. Horizontal Scaling (Scaling Out/In): Adding or removing VM instances to handle changes in workload demand.
   - Ideal for stateless applications or workloads with fluctuating demand.
   - `Virtual Machine Scale Sets (VMSS)`: Automatically scales the number of VMs based on demand.
   - `Azure Load Balancer`: Distributes traffic across multiple VMs.
   - Example: Scaling out from 2 VMs to 10 VMs during peak traffic and scaling back in during low traffic.
   - Limitations:
     - Requires applications to be stateless or use shared storage.
     - Increased complexity in managing multiple instances.
3. Scale Set(Virtual Machine Scale Sets - VMSS): A group of identical, load-balanced VMs that can automatically scale based on demand.
    - Designed for horizontal scaling.
    - Automatic scaling based on metrics (e.g., CPU, memory, or custom metrics).
    - High availability with distribution across Availability Zones or Availability Sets.
    - Integration with Azure Load Balancer or Application Gateway for traffic distribution.
    - Example: A web application experiences high traffic during business hours. VMSS automatically scales out from 5 VMs to 20 VMs during peak hours and scales back in during off-peak hours.
4. Availability Set: A logical grouping of VMs that ensures high availability by distributing VMs across.
   - Fault Domains: Isolated hardware to prevent single points of failure.
   - Update Domains: Groups of VMs that are updated sequentially to avoid downtime.
   - Protects against hardware failures and planned maintenance.
   - Does not provide automatic scaling.
   - Use-case: Running mission-critical applications that require high availability (e.g., databases, enterprise applications).
   - Example: Deploy 3 VMs in an Availability Set. If one VM fails or undergoes maintenance, the other VMs continue to run.
